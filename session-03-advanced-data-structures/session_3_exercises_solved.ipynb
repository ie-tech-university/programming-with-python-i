## Exercise 1: Student Grade Tracker
**Objective:** Practice lists, dictionaries, and the accumulation pattern.

Build a function that takes a list of student records (each record is a dictionary with `"name"` and `"scores"` keys) and returns a **report dictionary** mapping each student's name to their average score.

Then write a second function that takes that report and prints a formatted summary, including:
- Each student's name and average (rounded to 2 decimals)
- The class average
- The top-performing student

```python
def compute_averages(students):
    """Return a dict mapping each student name to their average score."""
    averages = {}
    for student in students:
        name = student["name"]
        scores = student["scores"]
        avg = sum(scores) / len(scores)
        averages[name] = round(avg, 2)
    return averages

def print_report(averages):
    """Print a formatted report with individual averages, class average, and top student."""
    print("\n===== GRADE REPORT =====")
    print(f"{'Student':<15} {'Average':>8}")
    print("-" * 25)

    for name, avg in averages.items():
        print(f"{name:<15} {avg:>8.2f}")

    class_avg = sum(averages.values()) / len(averages)
    print("-" * 25)
    print(f"{'Class Average':<15} {class_avg:>8.2f}")

    top_student = max(averages, key=averages.get)
    print(f"\n🏆 Top Student: {top_student} ({averages[top_student]})")

students = [
    {"name": "Alice", "scores": [90, 85, 92]},
    {"name": "Bob", "scores": [78, 81, 75]},
    {"name": "Charlie", "scores": [95, 98, 100]},
    {"name": "Diana", "scores": [88, 72, 91]},
]

averages = compute_averages(students)
print_report(averages)
```

## Exercise 2: Contact Book with Nested Dictionaries
**Objective:** Build a menu-driven application using nested dictionaries.

Create a contact book where each contact is stored as a **nested dictionary** inside a main dictionary. The key is the contact's name, and the value is another dictionary containing `"phone"` and `"email"`.

**Required features (menu loop):**
- **[A]dd** a new contact (name, phone, email). If the name already exists, ask whether to overwrite.
- **[S]earch** for a contact by name (case-insensitive).
- **[L]ist** all contacts in a formatted table.
- **[D]elete** a contact by name.
- **[Q]uit** the program.

```python
def run_contact_book():
    contacts = {}

    while True:
        print("\n===== CONTACT BOOK =====")
        print("[A]dd  [S]earch  [L]ist  [D]elete  [Q]uit")
        choice = input("> ").strip().lower()

        if choice == "a":
            name = input("Contact name: ").strip().title()
            if not name:
                print("Error: Name cannot be empty.")
                continue

            if name in contacts:
                overwrite = input(f"'{name}' already exists. Overwrite? (y/n): ").strip().lower()
                if overwrite != "y":
                    print("Cancelled.")
                    continue

            phone = input("Phone: ").strip()
            email = input("Email: ").strip()
            contacts[name] = {"phone": phone, "email": email}
            print(f"Contact '{name}' saved.")

        elif choice == "s":
            search = input("Search name: ").strip().title()
            if search in contacts:
                info = contacts[search]
                print(f"\n  Name:  {search}")
                print(f"  Phone: {info['phone']}")
                print(f"  Email: {info['email']}")
            else:
                print(f"Contact '{search}' not found.")

        elif choice == "l":
            if not contacts:
                print("Contact book is empty.")
            else:
                print(f"\n{'Name':<15} {'Phone':<15} {'Email'}")
                print("-" * 45)
                for name, info in contacts.items():
                    print(f"{name:<15} {info['phone']:<15} {info['email']}")

        elif choice == "d":
            name = input("Name to delete: ").strip().title()
            if name in contacts:
                del contacts[name]
                print(f"Contact '{name}' deleted.")
            else:
                print(f"Contact '{name}' not found.")

        elif choice == "q":
            print("Goodbye!")
            break

        else:
            print("Invalid option. Please try again.")

# run_contact_book()  # Uncomment to test
```

## Exercise 3: Company Data Parser (JSON-like Dict)
**Objective:** Practice navigating and extracting information from deeply nested data structures.

You are given a nested dictionary that represents a company's organizational structure (similar to what you'd receive from a JSON API). Write functions to extract specific information from it.

**Tasks:**
1. `get_all_employees(data)` → Return a flat list of **all** employee names across all departments.
2. `get_department_summary(data)` → Return a dict mapping each department name to its **employee count**.
3. `find_employee_department(data, employee_name)` → Return the name of the department an employee belongs to, or `"Not found"`.
4. `get_largest_department(data)` → Return the name of the department with the most employees.

```python
company_data = {
    "company": "TechCorp",
    "founded": 2010,
    "departments": [
        {
            "name": "Engineering",
            "manager": "Ava",
            "employees": ["Ava", "Liam", "Noah", "Emma", "Oliver"]
        },
        {
            "name": "Marketing",
            "manager": "Sophia",
            "employees": ["Sophia", "James", "Mia"]
        },
        {
            "name": "HR",
            "manager": "Isabella",
            "employees": ["Isabella", "Ethan"]
        },
        {
            "name": "Sales",
            "manager": "Lucas",
            "employees": ["Lucas", "Amelia", "Harper", "Ella"]
        }
    ]
}

def get_all_employees(data):
    """Return a flat list of all employee names."""
    all_employees = []
    for dept in data["departments"]:
        for emp in dept["employees"]:
            all_employees.append(emp)
    return all_employees

def get_department_summary(data):
    """Return a dict: {department_name: employee_count}."""
    summary = {}
    for dept in data["departments"]:
        summary[dept["name"]] = len(dept["employees"])
    return summary

def find_employee_department(data, employee_name):
    """Return the department name for the given employee, or 'Not found'."""
    for dept in data["departments"]:
        if employee_name in dept["employees"]:
            return dept["name"]
    return "Not found"

def get_largest_department(data):
    """Return the name of the department with the most employees."""
    summary = get_department_summary(data)
    return max(summary, key=summary.get)

# Test
print("All employees:", get_all_employees(company_data))
print("Summary:", get_department_summary(company_data))
print("Noah works in:", find_employee_department(company_data, "Noah"))
print("Zara works in:", find_employee_department(company_data, "Zara"))
print("Largest dept:", get_largest_department(company_data))
```

## Exercise 4: Inventory Deduplicator
**Objective:** Practice sets and tuples for deduplication and immutable records.

You receive a list of product shipments. Each shipment is a dictionary with `"product"`, `"sku"`, and `"quantity"`. Some shipments contain **duplicate SKUs** (the same product sent twice by mistake).

Write a function that:
1. Identifies and removes duplicate entries **by SKU** (keep the first occurrence).
2. Returns a tuple of: `(clean_list, duplicate_skus_set)`.
3. Prints a summary.

```python
def deduplicate_shipments(shipments):
    """
    Remove duplicate shipments by SKU (keep first occurrence).
    Return a tuple: (clean_list, duplicate_skus_set)
    """
    seen_skus = set()
    duplicate_skus = set()
    clean_list = []

    for item in shipments:
        sku = item["sku"]
        if sku in seen_skus:
            duplicate_skus.add(sku)
        else:
            seen_skus.add(sku)
            clean_list.append(item)

    return (clean_list, duplicate_skus)

shipments = [
    {"product": "Laptop", "sku": "LPT-001", "quantity": 10},
    {"product": "Mouse", "sku": "MSE-042", "quantity": 50},
    {"product": "Laptop", "sku": "LPT-001", "quantity": 10},
    {"product": "Keyboard", "sku": "KBD-007", "quantity": 30},
    {"product": "Monitor", "sku": "MNT-019", "quantity": 15},
    {"product": "Mouse", "sku": "MSE-042", "quantity": 50},
    {"product": "Webcam", "sku": "WBC-003", "quantity": 20},
    {"product": "Keyboard", "sku": "KBD-007", "quantity": 30},
]

clean, dupes = deduplicate_shipments(shipments)
print(f"\nReceived: {len(shipments)} shipments")
print(f"Unique: {len(clean)}")
print(f"Duplicates removed: {len(shipments) - len(clean)}")
print(f"Duplicated SKUs: {dupes}")
print(f"\nClean inventory:")
for item in clean:
    print(f"  {item['sku']}: {item['product']} (qty: {item['quantity']})")
```

## Exercise 5: Movie Collection Analyzer
**Objective:** Combine lists of dictionaries, sets, sorting, and filtering to analyze a dataset.

You have a list of movies, each represented as a dictionary. Write the following functions:

1. `get_unique_genres(movies)` → Return a **set** of all unique genres.
2. `movies_by_genre(movies, genre)` → Return a **list** of movie titles that belong to the given genre.
3. `average_rating(movies)` → Return the average rating across all movies.
4. `top_n_movies(movies, n)` → Return a **list** of the top `n` movies sorted by rating (highest first). Each item should be a tuple `(title, rating)`.
5. `director_filmography(movies)` → Return a **dict** mapping each director to a list of their movie titles.

```python
movies = [
    {"title": "Inception", "director": "Nolan", "genre": "Sci-Fi", "rating": 8.8},
    {"title": "The Dark Knight", "director": "Nolan", "genre": "Action", "rating": 9.0},
    {"title": "Pulp Fiction", "director": "Tarantino", "genre": "Crime", "rating": 8.9},
    {"title": "Interstellar", "director": "Nolan", "genre": "Sci-Fi", "rating": 8.6},
    {"title": "Kill Bill", "director": "Tarantino", "genre": "Action", "rating": 8.1},
    {"title": "Parasite", "director": "Bong", "genre": "Thriller", "rating": 8.5},
    {"title": "The Matrix", "director": "Wachowski", "genre": "Sci-Fi", "rating": 8.7},
    {"title": "Django Unchained", "director": "Tarantino", "genre": "Western", "rating": 8.4},
]

def get_unique_genres(movies):
    """Return a set of all unique genres."""
    genres = set()
    for movie in movies:
        genres.add(movie["genre"])
    return genres

def movies_by_genre(movies, genre):
    """Return a list of movie titles for the given genre."""
    result = []
    for movie in movies:
        if movie["genre"] == genre:
            result.append(movie["title"])
    return result

def average_rating(movies):
    """Return the average rating across all movies (rounded to 2 decimals)."""
    total = 0
    for movie in movies:
        total += movie["rating"]
    return round(total / len(movies), 2)

def top_n_movies(movies, n):
    """Return a list of (title, rating) tuples for the top n movies by rating."""
    sorted_movies = sorted(movies, key=lambda m: m["rating"], reverse=True)
    result = []
    for movie in sorted_movies[:n]:
        result.append((movie["title"], movie["rating"]))
    return result

def director_filmography(movies):
    """Return a dict mapping each director to a list of their movie titles."""
    filmography = {}
    for movie in movies:
        director = movie["director"]
        if director not in filmography:
            filmography[director] = []
        filmography[director].append(movie["title"])
    return filmography

# Test
print("Genres:", get_unique_genres(movies))
print("Sci-Fi movies:", movies_by_genre(movies, "Sci-Fi"))
print("Average rating:", average_rating(movies))
print("Top 3:", top_n_movies(movies, 3))
print("Filmography:", director_filmography(movies))
```

## Exercise 6: Event Planner (Capstone)
**Objective:** Combine **all** data structures (lists, dicts, tuples, sets) and Session II skills (menu loop, input validation, functions) into a single application.

Build a mini event planner where the user can:
- **[C]reate** an event (name, date, max capacity).
- **[R]egister** an attendee for an event (prevent duplicates using a set, enforce capacity).
- **[V]iew** event details: name, date, registered count / capacity, attendee list.
- **[S]ummary**: List all events with their registration counts.
- **[Q]uit**.

```python
def run_event_planner():
    events = {}

    while True:
        print("\n===== EVENT PLANNER =====")
        print("[C]reate  [R]egister  [V]iew  [S]ummary  [Q]uit")
        choice = input("> ").strip().lower()

        if choice == "c":
            name = input("Event name: ").strip()
            if not name:
                print("Error: Event name cannot be empty.")
                continue

            if name in events:
                print(f"Error: Event '{name}' already exists.")
                continue

            date = input("Date (YYYY-MM-DD): ").strip()

            try:
                capacity = int(input("Max capacity: "))
                if capacity <= 0:
                    print("Error: Capacity must be a positive number.")
                    continue
            except ValueError:
                print("Error: Capacity must be a whole number.")
                continue

            events[name] = {
                "date": date,
                "capacity": capacity,
                "attendees": set()
            }
            print(f"Event '{name}' created (capacity: {capacity}).")

        elif choice == "r":
            if not events:
                print("No events exist yet. Create one first.")
                continue

            event_name = input("Event name: ").strip()
            if event_name not in events:
                print(f"Error: Event '{event_name}' not found.")
                continue

            event = events[event_name]

            if len(event["attendees"]) >= event["capacity"]:
                print(f"Error: '{event_name}' is full ({event['capacity']}/{event['capacity']}).")
                continue

            email = input("Attendee email: ").strip().lower()
            if not email:
                print("Error: Email cannot be empty.")
                continue

            if email in event["attendees"]:
                print(f"'{email}' is already registered for this event.")
            else:
                event["attendees"].add(email)
                remaining = event["capacity"] - len(event["attendees"])
                print(f"'{email}' registered. {remaining} spot(s) remaining.")

        elif choice == "v":
            if not events:
                print("No events exist yet.")
                continue

            event_name = input("Event name: ").strip()
            if event_name not in events:
                print(f"Error: Event '{event_name}' not found.")
                continue

            event = events[event_name]
            print(f"\n{event_name}")
            print(f"   Date: {event['date']}")
            print(f"   Registered: {len(event['attendees'])} / {event['capacity']}")
            if event["attendees"]:
                print("   Attendees:")
                for attendee in sorted(event["attendees"]):
                    print(f"     - {attendee}")
            else:
                print("   No attendees registered yet.")

        elif choice == "s":
            if not events:
                print("No events exist yet.")
            else:
                print(f"\n{'Event':<25} {'Date':<15} {'Registered'}")
                print("-" * 50)
                for name, info in events.items():
                    reg = f"{len(info['attendees'])}/{info['capacity']}"
                    print(f"{name:<25} {info['date']:<15} {reg}")

        elif choice == "q":
            print("Goodbye!")
            break

        else:
            print("Invalid option. Please try again.")

# run_event_planner()  # Uncomment to test
```
