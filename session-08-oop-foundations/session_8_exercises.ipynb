## Exercise 1: Student Grade Tracker OOP
**Objective:** Build your first small object-oriented model using classes and methods.

Create a `Student` class and a `GradeBook` class to manage student records and calculate averages.

**Tasks:**
1. Create a `Student` class with:
   - `name`
   - `scores` (a list, default empty)
   - `add_score(score)` method
   - `average()` method
   - `letter_grade()` method
2. Create a `GradeBook` class with:
   - `students` list
   - `add_student(student)` method
   - `class_average()` method
   - `top_student()` method
3. Print a small report showing each student's average and letter grade.


```python
class Student:
    """Represent one student and their scores."""
    def __init__(self, name, scores=None):
        # TODO: store the student name
        # TODO: if scores is None, replace it with an empty list
        self.name = name
        self.scores = scores if scores is not None else []

    def add_score(self, score):
        """Add one score to the student's record."""
        # TODO
        self.scores.append(score)

    def average(self):
        """Return the average score, or 0 if no scores exist."""
        # TODO
        if self.scores is not None:
            return sum(self.scores)/len(self.scores)
        return 0

    def letter_grade(self):
        """Return A/B/C/D/F based on the average."""
        # TODO
        avg = self.average()

        if avg < 20:
            return "F"
        elif avg <= 40:
            return "D"
        elif avg <= 60:
            return "C"
        elif avg <= 80:
            return "B"
        return "A"


class GradeBook:
    """Store many Student objects and compute class statistics."""
    def __init__(self):
        # TODO: create an empty list of students
        self.students = []

    def add_student(self, student):
        """Append a Student object to the grade book."""
        # TODO
        self.students.append(student)

    def class_average(self):
        """Return the class average across all students."""
        # TODO
        self.students = [student for student in self.students if student.scores]
        if not self.students:
             return 0
        return sum([student.average() for student in self.students])/len(self.students)
             

    def top_student(self):
        """Return the Student object with the highest average."""
        # TODO
        if not self.students:
            return None
        return max(self.students, key=lambda student: student.average())


# --- TEST ---
alice = Student("Alice", [90, 85, 92])
bob = Student("Bob", [78, 81, 75])
charlie = Student("Charlie")
charlie.add_score(95)
charlie.add_score(98)

book = GradeBook()
for student in [alice, bob, charlie]:
    book.add_student(student)

for student in book.students:
    print(student.name, student.average(), student.letter_grade())

print("Class average:", round(book.class_average(), 2))
print("Top student:", book.top_student().name)
```

## Exercise 2: Contact Book Refactor
**Objective:** Refactor a contact book from nested dictionaries into classes.

Build:
- a `Contact` class for one person
- a `ContactBook` class to manage many contacts

**Tasks:**
1. Create a `Contact` class with `name`, `phone`, and `email`.
2. Add a `__str__()` method so contacts print nicely.
3. Create a `ContactBook` class that stores contacts in a dictionary using the lowercase name as the key.
4. Implement:
   - `add_contact(contact)`
   - `search(name)`
   - `delete(name)`
   - `list_contacts()`
5. Test your class methods without building the full menu loop yet.


```python
class Contact:
    """Represent a single contact."""
    def __init__(self, name, phone, email):
        # TODO
        self.name = name
        self.phone = phone
        self.email = email

    def __str__(self):
        # TODO: return a nicely formatted string
        return f"Name : {self.name}, phone number : {self.phone}, email : {self.email}"


class ContactBook:
    """Store and manage Contact objects."""
    def __init__(self):
        # TODO: create a dictionary to hold contacts
        self.contacts = []

    def add_contact(self, contact):
        # TODO: store by lowercase name
        if not contact:
            return ValueError("Bad input")
        name = contact.name.lower()
        self.contacts.append(name)

    def search(self, name):
        # TODO: return the matching Contact object or None
        name_lower = name.lower()
        for contact in self.contacts:
            if name_lower == contact:
                return name_lower
        return "None matched the input name"

    def delete(self, name):
        # TODO: remove and return True/False depending on success
        name_lower = name.lower()
        for contact in self.contacts:
            if name_lower == contact:
                self.contacts.remove(contact)
                return True
        return False
    
    def list_contacts(self):
        # TODO: return a list of Contact objects sorted by contact name
        self.contacts.sort()
        return self.contacts


# --- TEST ---
book = ContactBook()
book.add_contact(Contact("Alice", "555-1234", "alice@email.com"))
book.add_contact(Contact("Bob", "555-5678", "bob@email.com"))

print(book.search("alice"))
print(book.search("BOB"))
print(book.delete("Alice"))
print(book.delete("Charlie"))

for contact in book.list_contacts():
    print(contact)
```

## Exercise 3: Budget Ledger OOP
**Objective:** Model financial records using classes instead of loose dictionaries.

Create:
- a `Transaction` class
- a `BudgetLedger` class that stores transaction objects

**Tasks:**
1. Create a `Transaction` class with:
   - `amount`
   - `category`
   - `description`
2. Add a `__str__()` method for readable printing.
3. Create a `BudgetLedger` class with:
   - `transactions` list
   - `add_transaction(transaction)`
   - `total_spent()`
   - `totals_by_category()`
   - `most_expensive()`
4. Print a summary report.


```python
class Transaction:
    """Represent one expense transaction."""
    def __init__(self, amount=0, category= None, description= None):
        # TODO
        self.amount = amount
        self.category = category
        self.description = description

    def __str__(self):
        # TODO
        return f"Transaction (amount = {self.amount}), category = {self.category}, description = {self.description}"


class BudgetLedger:
    """Store many Transaction objects and summarize them."""
    def __init__(self):
        # TODO
        self.transactions = []

    def add_transaction(self, transaction):
        # TODO
        self.transactions.append(transaction)

    def total_spent(self):
        # TODO
        total = 0
        for transaction in self.transactions:
            total += transaction.amount
        return total
    
    def totals_by_category(self):
        # TODO: return a dictionary like {category: total}
        totals = {}
        for transaction in self.transactions:
            if transaction.category not in totals:
                totals[transaction.category] = 0    
            totals[transaction.category] += transaction.amount
        return totals
    
    def most_expensive(self):
        # TODO: return the Transaction object with the highest amount
        transaction_max = Transaction()
        max_value = 0
        for transaction in self.transactions:
            if transaction.amount > max_value:
                transaction_max = transaction
                max_value = transaction.amount
        return transaction_max
    
# --- TEST ---
ledger = BudgetLedger()
ledger.add_transaction(Transaction(12.50, "Food", "Lunch"))
ledger.add_transaction(Transaction(40.00, "Transport", "Train pass"))
ledger.add_transaction(Transaction(25.75, "Food", "Groceries"))

for t in ledger.transactions:
    print(t)

print("Total spent:", ledger.total_spent())
print("By category:", ledger.totals_by_category())
print("Most expensive:", ledger.most_expensive())
```

## Exercise 4: Procedural to OOP Refactor
**Objective:** Convert a procedural task manager into classes.

You are given a common procedural design: a list of dictionaries.
Refactor it into:
- a `Task` class
- a `TaskList` class

**Tasks:**
1. Create a `Task` class with:
   - `title`
   - `done` (default `False`)
   - `mark_done()`
2. Create a `TaskList` class with:
   - `tasks` list
   - `add_task(title)`
   - `pending_tasks()`
   - `completed_tasks()`
   - `mark_task_done(index)`
3. Print separate lists of pending and completed tasks.


```python
class Task:
    """Represent one task in a task list."""
    def __init__(self, title = "", done=False):
        # TODO
        self.title = title
        self.done = done

    def mark_done(self):
        # TODO
        self.done = True

    def __str__(self):
        # TODO: show status plus task title
        status = ""
        if self.done == True:
            status = "Done"
        else:
            status = "Incomplete"
        return print (f"Status is {status} and title : {self.title}")


class TaskList:
    """Store and manage Task objects."""
    def __init__(self):
        # TODO
        self.tasks = []

    def add_task(self, title):
        # TODO: create and append a Task object
        task = Task()
        task.title = title
        task.done = False
        self.tasks.append(task)

    def pending_tasks(self):
        # TODO
        pending_tasks = []
        for task in self.tasks:
            if task.done == False:
                pending_tasks.append(task)
        return pending_tasks
    
    def completed_tasks(self):
        # TODO
        completed_tasks = []
        for task in self.tasks:
            if task.done == True:
                completed_tasks.append(task)
        return completed_tasks

    def mark_task_done(self, index):
        # TODO: validate index and mark the correct task
        counter = 0
        for task in self.tasks:
            if counter == index:
                task.done = True
            counter += 1


# --- TEST ---
tasks = TaskList()
tasks.add_task("Read chapter 8")
tasks.add_task("Finish homework")
tasks.add_task("Review OOP notes")

tasks.mark_task_done(1)

print("Pending:")
for task in tasks.pending_tasks():
    print(" ", task.title)

print("Completed:")
for task in tasks.completed_tasks():
    print(" ", task.title)
```

## Exercise 5: Library Manager Capstone
**Objective:** Combine classes, encapsulation, and collections of objects in a larger mini-project.

Build:
- a `Book` class
- a `Library` class

**Tasks:**
1. `Book` should store:
   - `title`
   - `author`
   - `checked_out` (default `False`)
2. `Book` methods:
   - `checkout()`
   - `return_book()`
   - `__str__()`
3. `Library` should store a list of `Book` objects.
4. `Library` methods:
   - `add_book(book)`
   - `find_book(title)`
   - `available_books()`
   - `checked_out_books()`
5. Test checking out and returning books.


```python
class Book:
    """Represent one library book."""
    def __init__(self, title,author, checked_out = False, returned = False):
        # TODO
        self.title = title
        self.author = author
        self.checked_out = checked_out
        self.returned = returned

    def checkout(self):
        # TODO: mark as checked out if possible and return True/False
        self.checked_out = True

    def return_book(self):
        # TODO: mark as returned and return True/False
        if self.returned == False:
            self.returned = True
            return True
        return False

    def __str__(self):
        # TODO
        return print (f"The name of the book : {self.title}, author name : {self.author}, is it checked out :{self.checked_out}, is it returned : {self.returned}")


class Library:
    """Represent a small collection of books."""
    def __init__(self):
        # TODO
        self.books = []

    def add_book(self, book):
        # TODO
        return self.books.append(book)

    def find_book(self, title):
        # TODO: return matching Book object or None
        for book in self.books:
            if book.title == title:
                return book
        return None

    def available_books(self):
        # TODO
        available_books = []
        for book in self.books:
            if book.checked_out == False:
                available_books.append(book)
        return available_books
    
    def checked_out_books(self):
        # TODO
        checked_out_books = []
        for book in self.books:
            if book.checked_out == True:
                checked_out_books.append(book)
        return checked_out_books


# --- TEST ---
library = Library()
library.add_book(Book("1984", "George Orwell"))
library.add_book(Book("Dune", "Frank Herbert"))
library.add_book(Book("Pride and Prejudice", "Jane Austen"))

book = library.find_book("Dune")
if book:
    book.checkout()

print("Available books:")
for book in library.available_books():
    print(" ", book.title)

print("Checked out books:")
for book in library.checked_out_books():
    print(" ", book.title)
```
