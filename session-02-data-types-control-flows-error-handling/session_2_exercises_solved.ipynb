## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

```python
def clean_and_average(raw_data):
    valid_temps = []
    for item in raw_data:
        try:
            temp = float(item)
            valid_temps.append(temp)
        except (ValueError, TypeError):
            print(f"  Skipping invalid entry: {item!r}")
    
    if len(valid_temps) == 0:
        return 0.0
    
    average = sum(valid_temps) / len(valid_temps)
    return round(average, 2)

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings)}")
```

## Exercise 2: Transit Kiosk Menu Simulator
**Objective:** Build a menu-driven application with nested conditionals and defensive input validation.

Create a `while` loop that displays a menu for a transit ticket kiosk. Ask the user for their destination zone (1, 2, or 3) and their age. Apply a 50% discount if they are under 18 or over 65. 

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

```python
def run_kiosk():
    prices = {1: 2.50, 2: 4.00, 3: 6.50}

    while True:
        print("\n===== TRANSIT TICKET KIOSK =====")
        print("Zones: [1] $2.50  [2] $4.00  [3] $6.50")
        print("Type 'q' to quit.")

        zone_input = input("Select your zone (1/2/3): ")
        if zone_input.lower() == "q":
            print("Thank you. Goodbye!")
            break

        try:
            zone = int(zone_input)
            if zone not in prices:
                print("Error: Please select zone 1, 2, or 3.")
                continue
        except ValueError:
            print("Error: Zone must be a number (1, 2, or 3).")
            continue

        try:
            age = int(input("Enter your age: "))
            if age < 0 or age > 120:
                print("Error: Please enter a realistic age.")
                continue
        except ValueError:
            print("Error: Age must be a whole number.")
            continue

        base_price = prices[zone]
        if age < 18 or age > 65:
            final_price = base_price * 0.50
            discount_label = " (50% discount applied)"
        else:
            final_price = base_price
            discount_label = ""

        print(f"\n🎫 Zone {zone} ticket: ${final_price:.2f}{discount_label}")

run_kiosk()
```

## Exercise 3: Three-Strike Quiz System
**Objective:** Use loops, counters, and conditionals to manage state.

Write a short quiz program. The user must answer a question correctly. They have exactly 3 attempts. 
- If they get it right, print a success message and `break` out of the loop.
- If they fail 3 times, print a failure message.
- **Defensive Code:** Ensure you handle edge cases like different capitalizations or extra spaces in their input (e.g., " Paris " should be accepted).

```python
def run_quiz():
    question = "What is the capital of France?"
    correct_answer = "paris"
    max_attempts = 3
    attempts_used = 0

    print(f"Question: {question}")
    print(f"You have {max_attempts} attempts.\n")

    while attempts_used < max_attempts:
        user_answer = input(f"Attempt {attempts_used + 1}/{max_attempts}: ").strip().lower()
        attempts_used += 1

        if user_answer == correct_answer:
            print(f"✅ Correct! You got it in {attempts_used} attempt(s).")
            break
        else:
            remaining = max_attempts - attempts_used
            if remaining > 0:
                print(f"❌ Wrong. {remaining} attempt(s) remaining.")
            else:
                print(f"❌ Wrong. No attempts left. The answer was '{correct_answer.title()}'.")

run_quiz()
```

**Objective:** Identify repetitive code and refactor it into a reusable function with error handling.

The code below calculates the BMI (Body Mass Index) for three different patients. It is repetitive and lacks error handling. 

Refactor this by creating a `calculate_bmi(weight, height)` function that includes a `try/except` block to prevent `ZeroDivisionError` (if height is 0) or `TypeError` (if strings are passed).

```python
# --- BAD CODE (Repetitive and fragile) ---
# w1, h1 = 70, 1.75
# bmi1 = w1 / (h1 ** 2)
# print(f"Patient 1 BMI: {bmi1}")
#
# w2, h2 = 85, 1.80
# bmi2 = w2 / (h2 ** 2)
# print(f"Patient 2 BMI: {bmi2}")
#
# w3, h3 = "ninety", 0  # This will crash!
# bmi3 = w3 / (h3 ** 2)
# print(f"Patient 3 BMI: {bmi3}")

# --- REFACTORED CODE ---
def calculate_bmi(weight, height):
    """Calculate BMI given weight (kg) and height (m).
    Returns the BMI as a float rounded to 2 decimals, or None on error."""
    try:
        bmi = float(weight) / (float(height) ** 2)
        return round(bmi, 2)
    except ZeroDivisionError:
        print("  Error: Height cannot be zero.")
        return None
    except (TypeError, ValueError):
        print("  Error: Weight and height must be numeric values.")
        return None

patients = [
    ("Patient 1", 70, 1.75),
    ("Patient 2", 85, 1.80),
    ("Patient 3", "ninety", 0),
]

for name, w, h in patients:
    result = calculate_bmi(w, h)
    if result is not None:
        print(f"{name} BMI: {result}")
    else:
        print(f"{name} BMI: Could not be calculated.\n")
```

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

```python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    # First, convert from source unit to Celsius as the common base
    if from_unit == "C":
        celsius = value
    elif from_unit == "F":
        celsius = (value - 32) * 5 / 9
    elif from_unit == "K":
        celsius = value - 273.15
    else:
        raise ValueError(f"Unknown temperature unit: {from_unit}")

    # Validate: nothing can be below absolute zero
    if celsius < -273.15:
        raise ValueError("Temperature is below absolute zero (-273.15°C). Physically impossible.")

    # Convert from Celsius to target unit
    if to_unit == "C":
        return round(celsius, 2)
    elif to_unit == "F":
        return round(celsius * 9 / 5 + 32, 2)
    elif to_unit == "K":
        return round(celsius + 273.15, 2)
    else:
        raise ValueError(f"Unknown temperature unit: {to_unit}")


def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    if value < 0:
        raise ValueError("Distance cannot be negative.")

    # Convert everything to meters first
    to_meters = {"km": 1000, "miles": 1609.34, "m": 1}
    from_meters = {"km": 1 / 1000, "miles": 1 / 1609.34, "m": 1}

    if from_unit not in to_meters or to_unit not in from_meters:
        raise ValueError(f"Unknown distance unit. Choose from: km, miles, m")

    meters = value * to_meters[from_unit]
    result = meters * from_meters[to_unit]
    return round(result, 4)


def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    if value < 0:
        raise ValueError("Weight cannot be negative.")

    # Convert everything to grams first
    to_grams = {"kg": 1000, "lbs": 453.592, "g": 1}
    from_grams = {"kg": 1 / 1000, "lbs": 1 / 453.592, "g": 1}

    if from_unit not in to_grams or to_unit not in from_grams:
        raise ValueError(f"Unknown weight unit. Choose from: kg, lbs, g")

    grams = value * to_grams[from_unit]
    result = grams * from_grams[to_unit]
    return round(result, 4)


def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    categories = {
        "1": ("Temperature", convert_temperature, ["C", "F", "K"]),
        "2": ("Distance", convert_distance, ["km", "miles", "m"]),
        "3": ("Weight", convert_weight, ["kg", "lbs", "g"]),
    }

    while True:
        print("\n===== UNIVERSAL UNIT CONVERTER =====")
        print("[1] Temperature (°C, °F, K)")
        print("[2] Distance    (km, miles, m)")
        print("[3] Weight      (kg, lbs, g)")
        print("[Q] Quit")

        choice = input("\nSelect a category: ").strip().upper()

        if choice == "Q":
            print("Goodbye!")
            break

        if choice not in categories:
            print("Invalid choice. Please select 1, 2, 3, or Q.")
            continue

        name, converter_func, valid_units = categories[choice]
        print(f"\n--- {name} Converter ---")
        print(f"Available units: {', '.join(valid_units)}")

        try:
            value = float(input("Enter the value: "))
            from_unit = input(f"Convert FROM ({'/'.join(valid_units)}): ").strip()
            to_unit = input(f"Convert TO   ({'/'.join(valid_units)}): ").strip()

            result = converter_func(value, from_unit, to_unit)
            print(f"\n✅ {value} {from_unit} = {result} {to_unit}")

        except ValueError as e:
            print(f"\n❌ Input Error: {e}")

# run_converter()  # Uncomment to test
```

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

```python
def check_password_strength(password):
    """Evaluate password strength and return a score from 0 to 5."""
    score = 0
    feedback = []
    special_chars = "!@#$%^&*"

    # Rule 1: Length
    if len(password) >= 8:
        score += 1
    else:
        feedback.append("- Too short (need at least 8 characters)")

    # Rule 2: Uppercase
    has_upper = False
    for char in password:
        if char.isupper():
            has_upper = True
            break
    if has_upper:
        score += 1
    else:
        feedback.append("- Missing an uppercase letter")

    # Rule 3: Lowercase
    has_lower = False
    for char in password:
        if char.islower():
            has_lower = True
            break
    if has_lower:
        score += 1
    else:
        feedback.append("- Missing a lowercase letter")

    # Rule 4: Digit
    has_digit = False
    for char in password:
        if char.isdigit():
            has_digit = True
            break
    if has_digit:
        score += 1
    else:
        feedback.append("- Missing a digit")

    # Rule 5: Special character
    has_special = False
    for char in password:
        if char in special_chars:
            has_special = True
            break
    if has_special:
        score += 1
    else:
        feedback.append("- Missing a special character (!@#$%^&*)")

    return score, feedback


def run_password_checker():
    """Main loop: let the user test passwords until they type 'quit'."""
    strength_labels = {
        0: "💀 Very Weak",
        1: "😟 Weak",
        2: "😐 Fair",
        3: "🙂 Moderate",
        4: "💪 Strong",
        5: "🔒 Excellent",
    }

    print("===== PASSWORD STRENGTH CHECKER =====")
    print("Type 'quit' to exit.\n")

    while True:
        password = input("Enter a password to test: ")

        if password.strip().lower() == "quit":
            print("Goodbye!")
            break

        score, feedback = check_password_strength(password)
        print(f"\nScore: {score}/5 — {strength_labels[score]}")

        if feedback:
            print("Suggestions:")
            for item in feedback:
                print(f"  {item}")
        print()

# run_password_checker()  # Uncomment to test
```

## Exercise 7: Multi-Round Scoring Quiz
**Objective:** Combine loops, conditionals, lists, and dictionaries to build a complete quiz application with scoring.

Build a quiz that asks 5 questions. For each question, the user gets **2 attempts**. If they fail the first attempt, they receive a hint. Track their score and display a final report card at the end.

**Requirements:**
- Store questions in a list of dictionaries, each with keys: `question`, `answer`, `hint`.
- Clean user input (strip whitespace, convert to lowercase) before comparing.
- Award 2 points for a correct first attempt, 1 point for a correct second attempt, 0 for a miss.
- Display a final score and letter grade (A ≥ 9, B ≥ 7, C ≥ 5, else F).

```python
def run_scoring_quiz():
    """Run a multi-round quiz with hints and scoring."""
    questions = [
        {"question": "What planet is known as the Red Planet?", "answer": "mars", "hint": "It's the 4th planet from the Sun."},
        {"question": "What language is this course taught in?", "answer": "python", "hint": "It's named after a snake (sort of)."},
        {"question": "What does CPU stand for?", "answer": "central processing unit", "hint": "It's the 'brain' of the computer."},
        {"question": "What gas do plants absorb from the atmosphere?", "answer": "carbon dioxide", "hint": "Its chemical formula is CO2."},
        {"question": "In what city is IE University's main campus located?", "answer": "madrid", "hint": "It's the capital of Spain."},
    ]

    total_score = 0
    max_score = len(questions) * 2  # 2 points per question max
    results = []  # Store per-question results for the report

    print("===== MULTI-ROUND QUIZ =====")
    print(f"{len(questions)} questions. 2 attempts each. Good luck!\n")

    for i, q in enumerate(questions):
        print(f"Q{i + 1}: {q['question']}")
        points = 0

        # Attempt 1
        answer1 = input("  Your answer: ").strip().lower()
        if answer1 == q["answer"]:
            print("  ✅ Correct on first try! (+2 points)\n")
            points = 2
        else:
            # Attempt 2 with hint
            print(f"  ❌ Not quite. HINT: {q['hint']}")
            answer2 = input("  Try again: ").strip().lower()
            if answer2 == q["answer"]:
                print("  ✅ Correct on second try! (+1 point)\n")
                points = 1
            else:
                print(f"  ❌ The correct answer was: {q['answer'].title()}\n")
                points = 0

        total_score += points
        results.append({"question": q["question"], "points": points})

    # Final Report
    print("\n===== REPORT CARD =====")
    for i, r in enumerate(results):
        status = "✅" if r["points"] > 0 else "❌"
        print(f"  Q{i + 1}: {r['points']}/2 {status}  {r['question']}")

    print(f"\nTotal Score: {total_score}/{max_score}")

    # Letter grade using nested conditionals
    if total_score >= 9:
        grade = "A"
    elif total_score >= 7:
        grade = "B"
    elif total_score >= 5:
        grade = "C"
    else:
        grade = "F"

    print(f"Grade: {grade}")

# run_scoring_quiz()  # Uncomment to test
```

## Exercise 8: Expense Tracker with Category Reports
**Objective:** Capstone exercise combining **all** Session II concepts: menu-driven loops, nested conditionals, `try/except`, type conversion, functions, and dictionary aggregation.

Build a mini expense tracker where the user can:
1. **Add an expense** — enter a category (e.g., Food, Transport), a description, and an amount.
2. **View all expenses** — display a numbered list of all recorded expenses.
3. **View summary by category** — show the total spent per category and the percentage of overall spending.
4. **Quit** — exit the program.

**Requirements:**
- Use `try/except` to reject non-numeric amounts.
- Use conditionals to reject negative amounts.
- Normalize category names (capitalize the first letter).
- Use a function for each menu action.
- Handle division by zero if the user requests a summary with no expenses.

```python
def add_expense(expenses):
    """Prompt the user for expense details and append to the list."""
    category = input("  Category (e.g., Food, Transport, Fun): ").strip().capitalize()
    if not category:
        print("  ❌ Category cannot be empty.")
        return

    description = input("  Description: ").strip()
    if not description:
        print("  ❌ Description cannot be empty.")
        return

    try:
        amount = float(input("  Amount ($): "))
        if amount <= 0:
            print("  ❌ Amount must be a positive number.")
            return
    except ValueError:
        print("  ❌ Invalid amount. Please enter a number.")
        return

    expenses.append({
        "category": category,
        "description": description,
        "amount": amount,
    })
    print(f"  ✅ Added: {description} ({category}) — ${amount:.2f}")


def view_expenses(expenses):
    """Display all recorded expenses in a formatted list."""
    if len(expenses) == 0:
        print("  No expenses recorded yet.")
        return

    print(f"\n  {'#':<4} {'Category':<14} {'Description':<22} {'Amount':>8}")
    print("  " + "-" * 50)
    for i, exp in enumerate(expenses):
        print(f"  {i+1:<4} {exp['category']:<14} {exp['description']:<22} ${exp['amount']:>7.2f}")

    total = sum(exp["amount"] for exp in expenses)
    print("  " + "-" * 50)
    print(f"  {'TOTAL':<40} ${total:>7.2f}\n")


def view_summary(expenses):
    """Show spending totals and percentages grouped by category."""
    if len(expenses) == 0:
        print("  No expenses recorded yet.")
        return

    # Aggregate by category using a dictionary
    category_totals = {}
    for exp in expenses:
        cat = exp["category"]
        if cat in category_totals:
            category_totals[cat] += exp["amount"]
        else:
            category_totals[cat] = exp["amount"]

    grand_total = sum(category_totals.values())

    print(f"\n  {'Category':<14} {'Total':>10} {'Pct':>8}")
    print("  " + "-" * 34)

    for cat, total in category_totals.items():
        try:
            pct = (total / grand_total) * 100
        except ZeroDivisionError:
            pct = 0.0
        print(f"  {cat:<14} ${total:>9.2f} {pct:>7.1f}%")

    print("  " + "-" * 34)
    print(f"  {'GRAND TOTAL':<14} ${grand_total:>9.2f} {100.0:>7.1f}%\n")


def run_expense_tracker():
    """Main menu loop for the expense tracker."""
    expenses = []

    while True:
        print("\n===== EXPENSE TRACKER =====")
        print("[1] Add an expense")
        print("[2] View all expenses")
        print("[3] View summary by category")
        print("[Q] Quit")

        choice = input("\nSelect an option: ").strip().upper()

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_expenses(expenses)
        elif choice == "3":
            view_summary(expenses)
        elif choice == "Q":
            print("Goodbye!")
            break
        else:
            print("❌ Invalid option. Please choose 1, 2, 3, or Q.")

# run_expense_tracker()  # Uncomment to test
```
