## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data: list):
    valid_temps = []
    for item in raw_data:
        try:
            valid_temps.append(float(item))
        except:
            pass

    if len(valid_temps) != 0:
        return sum(valid_temps)/len(valid_temps)
    else:
        return 0

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings):.2f}")
````

## Exercise 2: Transit Kiosk Menu Simulator
**Objective:** Build a menu-driven application with nested conditionals and defensive input validation.

Create a `while` loop that displays a menu for a transit ticket kiosk. Ask the user for their destination zone (1, 2, or 3) and their age. Apply a 50% discount if they are under 18 or over 65. 

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

````python
def run_kiosk():
    while True:
        try:
            zone = int(input("Zone: 1 (€2), 2 (€3), 3 (€4): "))
            age = int(input("Age: "))
            if zone not in (1, 2, 3) or age < 0:
                raise ValueError
            price = zone + 1
            if age < 18 or age > 65:
                price /= 2
            print(f"Ticket: €{price:.2f}")
            if input("Another ticket? (y/n): ").lower() != "y":
                break
        except ValueError:
            print("Invalid input. Please try again.")

run_kiosk()
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    if from_unit == "celsius":
        c = value
    elif from_unit == "fahrenheit":
        c = (value - 32) * 5 / 9
    else:
        c = value - 273.15

    if c < -273.15:
        raise ValueError("Below absolute zero.")
    
    if to_unit == "celsius":
        return c
    elif to_unit == "fahrenheit":
        return c * 9 / 5 + 32
    else:
        return c + 273.15


def convert_distance(value, from_unit, to_unit):
    factors = {"km": 1000, "miles": 1609.344, "meters": 1}
    return convert(value, from_unit, to_unit, factors)


def convert_weight(value, from_unit, to_unit):
    factors = {"kg": 1000, "lbs": 453.59237, "grams": 1}
    return convert(value, from_unit, to_unit, factors)


def convert(value, from_unit, to_unit, factors):
    if value < 0:
        raise ValueError("Value cannot be negative.")
    return value * factors[from_unit] / factors[to_unit]


def run_converter():
    options = {"1": ("Temperature", convert_temperature),
               "2": ("Distance", convert_distance),
               "3": ("Weight", convert_weight)}

    while True:
        choice = input("\n[1] Temperature [2] Distance [3] Weight [0] Quit: ")
        if choice == "0" or choice.lower() == "quit":
            break
        try:
            _, function = options[choice]
            value = float(input("Value: "))
            from_unit = input("From unit: ").lower()
            to_unit = input("To unit: ").lower()
            print(function(value, from_unit, to_unit), to_unit)
        except (ValueError, KeyError) as error:
            print("Invalid input:", error)


run_converter()
````

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

**Output:**
- 0–1: `Weak`
- 2–3: `Medium`
- 4–5: `Strong`

**Hint:** Use a `for` loop to iterate through each character of the password string and check its properties with `.isupper()`, `.isdigit()`, etc.

````python
def check_password_strength(password):
    """Evaluate a password and return a score (0-5) and a label."""
    score = 0
    special_chars = "!@#$%^&*"

    # I use one flag per rule, all False at the start
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False

    # Rule 1: length at least 8 characters
    if len(password) >= 8:
        score += 1

    # I loop through each character and turn on the matching flag
    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True
        elif char.isdigit():
            has_digit = True
        elif char in special_chars:
            has_special = True

    # Rules 2 to 5: each flag that is True adds 1 point
    if has_upper:
        score += 1
    if has_lower:
        score += 1
    if has_digit:
        score += 1
    if has_special:
        score += 1

    # I turn the score into a label
    if score <= 1:
        label = "Weak"
    elif score <= 3:
        label = "Medium"
    else:
        label = "Strong"

    return score, label


def run_password_checker():
    """Loop: ask user for passwords until they type 'quit'."""
    while True:
        password = input("Enter a password (or type 'quit' to exit): ")

        # I only strip spaces to check for quit, not the password itself,
        # because spaces could be part of a real password
        if password.strip().lower() == "quit":
            print("Goodbye!")
            break

        score, label = check_password_strength(password)
        print(f"Score: {score}/5 - {label}")


run_password_checker()
````

## Exercise 7: Multi-Round Scoring Quiz (with Data Structures)
**Objective:** Combine loops, conditionals, error handling, and dictionaries to build a stateful application.

Upgrade Exercise 3 into a full multi-question quiz. The questions and answers are stored in a **list of dictionaries**. The program tracks the user's score across all questions and displays a final report card.

**Requirements:**
- Store at least 5 questions in a list of dicts: `[{"question": ..., "answer": ..., "hint": ...}, ...]`
- The user gets **2 attempts per question**. After the first wrong answer, display the hint.
- Clean all user input (`.strip().lower()`).
- At the end, print a summary: total score, percentage, and a grade (A/B/C/F).
- Use `try/except` if any question expects a numeric answer.

````python
quiz_bank = [
    {"question": "What keyword creates a function in Python?", "answer": "def", "hint": "It's a 3-letter word."},
    {"question": "What data type is immutable: list or tuple?", "answer": "tuple", "hint": "It uses parentheses."},
    {"question": "What is 2 ** 10?", "answer": "1024", "hint": "It's a power of 2."},
    {"question": "What method removes whitespace from both ends of a string?", "answer": "strip", "hint": "Think .___()."},
    {"question": "What built-in function returns the number of items in a list?", "answer": "len", "hint": "Short for length."},
]

def run_full_quiz(questions):
    score = 0
    total = len(questions)

    # I loop through each question dictionary, numbering them from 1
    for number, item in enumerate(questions, start=1):
        print(f"\nQuestion {number} of {total}")
        correct = False

        # The user gets 2 attempts per question
        for attempt in range(1, 3):
            answer = input(item["question"] + " ")

            # I clean the input so " TUPLE " is accepted
            answer = answer.strip().lower()

            if answer == item["answer"]:
                print("Correct!")
                score += 1
                correct = True
                break   # no need for a second attempt

            # Wrong answer: on the first attempt I show the hint
            if attempt == 1:
                print("Wrong. Hint:", item["hint"])
            else:
                print(f"Wrong again. The answer was: {item['answer']}")

        # (the "correct" flag isn't needed for scoring, but it makes my logic clear)

    # I calculate the percentage once all questions are done
    percentage = score / total * 100

    # I turn the percentage into a letter grade
    if percentage >= 100:
        grade = "A"
    elif percentage >= 80:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    else:
        grade = "F"

    # Final report card
    print("\n--- REPORT CARD ---")
    print(f"Score: {score}/{total}")
    print(f"Percentage: {percentage:.1f}%")
    print(f"Grade: {grade}")


run_full_quiz(quiz_bank)
````

## Exercise 8: Expense Tracker with Category Reports
**Objective:** Build a complete menu-driven application that combines everything: loops, nested conditionals, error handling, functions, and data structures.

Create a mini expense tracker where the user can add expenses, view them by category, and see a summary report. This exercise ties together all the concepts from Session II.

**Menu Options:**
1. **Add Expense** — Ask for amount (float), category (string), and description (string). Validate that the amount is a positive number.
2. **View by Category** — Show all expenses for a user-selected category.
3. **Summary Report** — Print: total spent, number of transactions, average transaction, and the category with the highest spend.
4. **Quit**

**Data Structure:** Store expenses as a list of dictionaries:
```python
[{"amount": 12.50, "category": "food", "description": "Lunch"}, ...]
```

**Defensive Coding Requirements:**
- `try/except ValueError` for the amount input.
- Reject negative or zero amounts.
- Handle the case where the user asks for a summary with zero expenses (avoid `ZeroDivisionError`).

````python
expenses = []

def add_expense(expenses_list):
    """Prompt user for expense details and append to the list."""
    # I ask for the amount first, and try/except catches letters
    try:
        amount = float(input("Amount: "))
    except ValueError:
        print("Error: please enter a number, like 12.50.")
        return

    # I reject zero and negative amounts
    if amount <= 0:
        print("Error: the amount must be greater than zero.")
        return

    # I clean the category so "Food" and " food " count as the same category
    category = input("Category: ").strip().lower()
    description = input("Description: ").strip()

    # I don't accept an empty category, because it would break the reports
    if category == "":
        print("Error: the category cannot be empty.")
        return

    # Each expense is a dictionary, as the exercise asks
    expenses_list.append({
        "amount": amount,
        "category": category,
        "description": description,
    })
    print("Expense added.")


def view_by_category(expenses_list):
    """Show expenses filtered by a user-selected category."""
    if not expenses_list:
        print("No expenses yet.")
        return

    # A set gives me each category only once, and sorted() puts them in order
    categories = sorted(set(item["category"] for item in expenses_list))
    print("Categories:", ", ".join(categories))

    choice = input("Which category do you want to see? ").strip().lower()

    # I loop and print only the matching expenses
    found = False
    for item in expenses_list:
        if item["category"] == choice:
            print(f"- {item['amount']:.2f} | {item['description']}")
            found = True

    if not found:
        print("No expenses found for that category.")


def summary_report(expenses_list):
    """Print total, count, average, and top category."""
    # Empty list: I stop here so there is no ZeroDivisionError
    if not expenses_list:
        print("No expenses yet, so there is nothing to summarize.")
        return

    count = len(expenses_list)
    total = 0
    totals_by_category = {}   # category -> total spent in it

    # I add up the overall total and the total of each category
    for item in expenses_list:
        total += item["amount"]
        category = item["category"]
        totals_by_category[category] = totals_by_category.get(category, 0) + item["amount"]

    average = total / count

    # max() with key= gives me the category with the highest total
    top_category = max(totals_by_category, key=totals_by_category.get)

    print("\n--- SUMMARY REPORT ---")
    print(f"Total spent: {total:.2f}")
    print(f"Number of transactions: {count}")
    print(f"Average transaction: {average:.2f}")
    print(f"Top category: {top_category} ({totals_by_category[top_category]:.2f})")


def run_tracker():
    """Main menu loop."""
    while True:
        print("\n--- EXPENSE TRACKER ---")
        print("[1] Add Expense")
        print("[2] View by Category")
        print("[3] Summary Report")
        print("[4] Quit")
        choice = input("Choose an option: ").strip()

        # I route each choice to the right function
        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_by_category(expenses)
        elif choice == "3":
            summary_report(expenses)
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            # Invalid menu choice: friendly message, then the menu shows again
            print("Please choose 1, 2, 3 or 4.")


run_tracker()
````
