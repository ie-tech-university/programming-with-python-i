## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data):
    valid_temps = []
    # TODO: Loop through raw_data
    # TODO: Use try/except to convert each item to a float
    # TODO: Append valid floats to valid_temps
    # TODO: Calculate and return the average
    for data in raw_data:
        try:
            temp = float(data)
            valid_temps.append(temp)
        except (ValueError, TypeError):
            continue
    
    if not valid_temps:
        return 0

    average = sum(valid_temps) / len(valid_temps)
    
    return average

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings)}")
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    # TODO: Raise ValueError if result is below absolute zero
    # TODO: Implement conversion logic
    if from_unit == "C":
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is not possible.")
        if to_unit == "F":
            result = (value * 9/5) + 32
        else:
            result = value + 273.15
    elif from_unit == "F":
        if value < -459.67:
            raise ValueError("Temperature below absolute zero is not possible.")
            
        if to_unit == "C":
            result = (value - 32) * 5/9
        else:
            result = (value + 459.67) * 5/9
    else: # from_unit == "K"
        if value < 0:
            raise ValueError("Temperature below absolute zero is not possible.")
        if to_unit == "C":
            result = value - 273.15
        else:
            result = (value * 9/5) - 459.67
    return result   


def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    # TODO: Implement conversion logic
    # TODO: Reject negative distances
    if value < 0:
        raise ValueError("Distance cannot be negative.")
    if from_unit == "km":
        if to_unit == "miles":
            result = value * 0.621371
        else:
            result = value * 1000
    elif from_unit == "miles":
        if to_unit == "km":
            result = value * 1.60934
        else:
            result = value * 1609.34
    else: # from_unit == "m"
        if to_unit == "km":
            result = value / 1000
        else:
            result = value / 1609.34
    return result



def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    # TODO: Implement conversion logic
    # TODO: Reject negative weights
    if value < 0:
        raise ValueError("Weight cannot be negative.")
    if from_unit == "kg":
        if to_unit == "lbs":
            result = value * 2.20462
        else:
            result = value * 1000
    elif from_unit == "lbs":
        if to_unit == "kg":
            result = value / 2.20462
        else:
            result = value * 453.592
    else: # from_unit == "g"
        if to_unit == "kg":
            result = value / 1000
        else:
            result = value / 453.592
    return result

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    # TODO: while True loop with menu: [1] Temperature [2] Distance [3] Weight [Q] Quit
    # TODO: Inside each option, ask for value, from_unit, to_unit
    # TODO: Wrap everything in try/except to catch ValueError and bad input
    while True:
        print("Converter Menu:")
        print("[1] Temperature")
        print("[2] Distance")
        print("[3] Weight")
        print("[Q] Quit")
        valid_choices = ['1', '2', '3', 'Q']
        choice = input("Select an option: (1/2/3/Q)").strip().upper()
        
        if choice not in valid_choices:
            print("Invalid option. Please select 1, 2, 3, or Q.")
            continue
        else:
            if choice == 'Q':
                break
            elif choice == '1':
                valid_units = ['C', 'F', 'K']
                print("Temperature conversion selected. Supported units: C, F, K")
            elif choice == '2':
                valid_units = ['m', 'km', 'miles']
                print("Distance conversion selected. Supported units: m, km, miles")
            elif choice == '3':
                valid_units = ['kg', 'lbs', 'g']
                print("Weight conversion selected. Supported units: kg, lbs, g")
            
            while True:
                from_unit = input("Enter the unit to convert from: ").strip().lower()
                if from_unit not in valid_units:
                    print(f"Invalid unit. Please select from {valid_units}.")
                    continue
                to_unit = input("Enter the unit to convert to: ").strip().lower()
                if to_unit not in valid_units:
                    print(f"Invalid unit. Please select from {valid_units}.")
                    continue
                if from_unit == to_unit:
                    print("From and To units cannot be the same. Please choose different units.")
                    continue
                break
            while True:
                try:
                    value = float(input("Enter the value to convert: "))
                    if choice == '1':
                        result = convert_temperature(value, from_unit, to_unit)
                    elif choice == '2':
                        result = convert_distance(value, from_unit, to_unit)
                    elif choice == '3':
                        result = convert_weight(value, from_unit, to_unit)
                    print(f"Converted value: {result} {to_unit}")
                    break
                except ValueError:
                    continue

                
run_converter()  # Uncomment to test
````

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

**Output:**
- 0–1: `Weak`
- 2–3: `Medium`
- 4–5: `Strong`

**Hint:** Use a `for` loop to iterate through each character of the password string and check its properties with `.isupper()`, `.isdigit()`, etc.

````python
def check_password_strength(password):
    """Evaluate a password and return a score (0-5) and a label."""
    score = 0
    special_chars = "!@#$%^&*"
    length = 0
    upper = 0
    lower = 0
    digit = 0
    special = 0
    # TODO: Check length >= 8
    # TODO: Loop through characters to check for uppercase, lowercase, digit, special
    # TODO: Use boolean flags (has_upper, has_lower, etc.) and add to score
    # TODO: Return score and strength label
    if len(password) >= 8:
        length += 1
    for characters in password:
        if characters.isupper():
            upper += 1
        if characters.islower():
            lower += 1
        if characters.isdigit():
            digit += 1
        for special_char in special_chars:
            if characters == special_char:
                special += 1

    checks = [length, upper, lower, digit, special]
    for check in checks:
        if check > 0:
            score += 1
    return score
    

def run_password_checker():
    """Loop: ask user for passwords until they type 'quit'."""
    # TODO: while loop, get input, call check_password_strength, print result
    while True:
        user_input = input("Please enter your password: (type 'quit' to exit)").strip()
        if user_input == 'quit':
            break

        # **Output:**
        # - 0–1: `Weak`
        # - 2–3: `Medium`
        # - 4–5: `Strong`
        score = check_password_strength(user_input)
        if score < 2:
            print("Weak")
        elif score < 4:
            print("Medium")
        else: #score 4-5
            print("Strong")



run_password_checker()  # Uncomment to test
````

````python
expenses = []

# test_expenses = [
#     {"amount": 12.50, "category": "food", "description": "Lunch"},
#     {"amount": 3.20, "category": "transport", "description": "Bus fare"},
#     {"amount": 45.00, "category": "shopping", "description": "T-shirt"},
#     {"amount": 8.75, "category": "food", "description": "Dinner"},
#     {"amount": 2.50, "category": "transport", "description": "Metro ticket"},
#     {"amount": 60.00, "category": "entertainment", "description": "Concert ticket"},
#     {"amount": 15.30, "category": "food", "description": "Groceries"},
#     {"amount": 25.00, "category": "utilities", "description": "Phone bill"},
#     {"amount": 0.99, "category": "shopping", "description": "Pen"},
#     {"amount": 120.00, "category": "travel", "description": "Hotel"}
# ]

def add_expense(expenses_list):
    """Prompt user for expense details and append to the list."""
    # TODO: Ask for amount with try/except for ValueError
    # TODO: Reject negative or zero amounts
    # TODO: Ask for category and description
    # TODO: Append a dict to expenses_list
    while True:
        try:
            amount = float(input("Enter the expense amount: "))
            if amount <= 0:
                print("Amount must be greater than zero. Please try again.")
                continue
            category = input("Enter the expense category: ").strip()
            description = input("Enter a brief description of the expense: ").strip()
            expenses_list.append({"amount": amount, "category": category, "description": description})
            print(f"Expense added: {amount} in category '{category}' with description '{description}'.")
            break
        except ValueError:
            print("Invalid input. Please enter a numeric value for the amount.")

def view_by_category(expenses_list):
    """Show expenses filtered by a user-selected category."""
    # TODO: Get unique categories from the list
    # TODO: Ask user to pick one
    # TODO: Loop and print matching expenses
    if not expenses_list:
        print("No expenses recorded yet.")
        return
    unique_categories = set(expense["category"] for expense in expenses_list)
    print("Available categories:")
    for i, category in enumerate(unique_categories, 1):
        print(f"{i}. {category}")
    while True:
        try:
            choice = int(input("Select a category by number: "))
            if 1 <= choice <= len(unique_categories):
                selected_category = list(unique_categories)[choice - 1]
                print(f"Expenses in category '{selected_category}':")
                for expense in expenses_list:
                    if expense["category"] == selected_category:
                        print(f"  - {expense['amount']}: {expense['description']}")
                break
            else:
                print("Invalid choice. Please select a valid category.")
        except ValueError:
            print("Invalid input. Please enter a number.")

def summary_report(expenses_list):
    total = 0
    count = 0
    category_totals = {}
    """Print total, count, average, and top category."""
    # TODO: Handle empty list case (no ZeroDivisionError!)
    # TODO: Calculate total, count, average
    # TODO: Find the category with the highest total spend
    if not expenses_list:
        print("No expenses recorded yet.")
        return
    for expense in expenses_list:
        total += expense["amount"]
        count += 1
        category_totals[expense["category"]] = category_totals.get(expense["category"], 0) + expense["amount"]

    average = round(total/count, 2)
    top_category = max(category_totals, key=category_totals.get)
    top_amount = round(category_totals[top_category], 2)
    print(f"Total expenses: {total}")
    print(f"Number of expenses: {count}")
    print(f"Average expense: {average}")
    print(f"Top category: {top_category.upper()} with total spend: {top_amount}")

def run_tracker():
    """Main menu loop."""
    # TODO: while True with menu: [1] Add [2] View by Category [3] Summary [4] Quit
    # TODO: Use conditionals to route to the right function
    # TODO: Handle invalid menu choices gracefully
    while True:
        print("Expense Tracker Menu:")
        print("[1] Add")
        print("[2] View by Category")
        print("[3] Summary")
        print("[4] Quit")
        valid_choices = ['1', '2', '3', '4']
        choice = str(input("Select an option: (1/2/3/4)").strip())
        if choice not in valid_choices:
            print("Invalid option. Please select 1, 2, 3, or 4.")
            continue
        else:
            if choice == '4':
                break
            elif choice == '1':
                add_expense(expenses)
            elif choice == '2':
                view_by_category(expenses)
            elif choice == '3':
                summary_report(expenses)
        
run_tracker()  # Uncomment to test
````
