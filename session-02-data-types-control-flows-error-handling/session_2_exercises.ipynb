## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data):
    valid_temps = []
    for data in raw_data:
        if data is not None:
            try:
                valid_temp = float(data)
                valid_temps.append(valid_temp)
            except ValueError:
                print(f"Data entry in raw data : {data} can't be turned into a float value")
                continue
    if not valid_temps:
        return 0.0
    return sum(valid_temps)/len(valid_temps)
    # TODO: Loop through raw_data
    # TODO: Use try/except to convert each item to a float
    # TODO: Append valid floats to valid_temps
    # TODO: Calculate and return the average

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings)}")
````

**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    if from_unit == "C":
        if to_unit == "F":
            result = (value * 9/5) + 32
        elif to_unit == "K":
            result = value + 273.15
        else:
            result = value
    elif from_unit == "F":  
        if to_unit == "C":
            result = (value - 32) * 5/9
        elif to_unit == "K":
            result = (value - 32) * 5/9 + 273.15
        else:
            result = value
    elif from_unit == "K":  
        if to_unit == "C":
            result = value - 273.15
        elif to_unit == "F":
            result = (value - 273.15) * 9/5 + 32
        else:
            result = value
    else:
        raise ValueError(f"Invalid from_unit: {from_unit}. Must be 'C', 'F', or 'K'.")  
    # TODO: Implement conversion logic
    # TODO: Raise ValueError if result is below absolute zero
    if result < -273.15 and to_unit == "K":
        raise ValueError(f"Invalid value: {value}. Must be a positive number.")
    return result

def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    if from_unit == "km":
        if to_unit == "miles":
            result = value * 0.621371
        elif to_unit == "m":
            result = value * 1000
        else:
            result = value
    elif from_unit == "miles":
        if to_unit == "km":
            result = value / 0.621371
        elif to_unit == "m":
            result = value * 1609.34
        else:
            result = value
    elif from_unit == "m":
        if to_unit == "km":
            result = value / 1000
        elif to_unit == "miles":
            result = value / 1609.34
        else:
            result = value
    else:
        raise ValueError(f"Invalid from_unit: {from_unit}. Must be 'km', 'miles', or 'm'.")
    # TODO: Implement conversion logic
    # TODO: Reject negative distances
    if result < 0:
        raise ValueError(f"Invalid value: {value}. Must be a positive number.")
    return result

def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    if from_unit == "kg":
        if to_unit == "lbs":
            result = value * 2.20462
        elif to_unit == "g":
            result = value * 1000
        else:
            result = value
    elif from_unit == "lbs":
        if to_unit == "kg":
            result = value / 2.20462
        elif to_unit == "g":
            result = value * 453.592
        else:
            result = value
    elif from_unit == "g":
        if to_unit == "kg":
            result = value / 1000
        elif to_unit == "lbs":
            result = value / 453.592
        else:
            result = value
    else:
        raise ValueError(f"Invalid from_unit: {from_unit}. Must be 'kg', 'lbs', or 'g'.")
    # TODO: Implement conversion logic
    # TODO: Reject negative weights
    if result < 0:
        raise ValueError(f"Invalid value: {value}. Must be a positive number.")
    return result

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    while True:
        print("Unit Converter Menu:")
        print("[1] Temperature")
        print("[2] Distance")
        print("[3] Weight")
        print("[Q] Quit")
        choice = input("Please select an option: ").strip().upper()

        if choice == "Q":
            print("Exiting the converter. Goodbye!")
            break
        elif choice in ["1", "2", "3"]:
            try:
                value = float(input("Enter the value to convert: "))
                from_unit = input("Enter the from unit: ").strip()
                to_unit = input("Enter the to unit: ").strip()

                if choice == "1":
                    result = convert_temperature(value, from_unit, to_unit)
                elif choice == "2":
                    result = convert_distance(value, from_unit, to_unit)
                elif choice == "3":
                    result = convert_weight(value, from_unit, to_unit)

                print(f"Converted value: {result}")
            except ValueError as e:
                print(f"Error: {e}")
            except Exception as e:
                print(f"An unexpected error occurred: {e}")
        else:
            print("Invalid option. Please try again.")
    # TODO: while True loop with menu: [1] Temperature [2] Distance [3] Weight [Q] Quit
    # TODO: Inside each option, ask for value, from_unit, to_unit
    # TODO: Wrap everything in try/except to catch ValueError and bad input
run_converter()  # Uncomment to test
````

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

**Output:**
- 0–1: `Weak`
- 2–3: `Medium`
- 4–5: `Strong`

**Hint:** Use a `for` loop to iterate through each character of the password string and check its properties with `.isupper()`, `.isdigit()`, etc.

````python
def check_password_strength(password):
    """Evaluate a password and return a score (0-5) and a label."""
    score = 0
    special_chars = "!@#$%^&*"
    if len(password) >= 8:
        score += 1
    # TODO: Check length >= 8
    # TODO: Loop through characters to check for uppercase, lowercase, digit, special
    for char in password:
        if char.isupper():
            score += 1
            break
    # TODO: Use boolean flags (has_upper, has_lower, etc.) and add to score
    if any(char.islower() for char in password):
        score += 1
    if any(char.isdigit() for char in password):
        score += 1
    if any(char in special_chars for char in password):
        score += 1
    # TODO: Return score and strength label
    if score == 0:
        label = "Very Weak"
    elif score == 1:
        label = "Weak"
    elif score == 2:
        label = "Moderate"
    elif score == 3:
        label = "Strong"
    else:
        label = "Very Strong"
    return score, label

def run_password_checker():
    """Loop: ask user for passwords until they type 'quit'."""
    # TODO: while loop, get input, call check_password_strength, print result
    while True:
        password = input("Enter a password to check (or type 'quit' to exit): ")
        if password.lower() == 'quit':
            print("Exiting password checker.")
            break
        score, label = check_password_strength(password)
        print(f"Password strength score: {score}, Label: {label}")

run_password_checker()  # Uncomment to test
````

## Exercise 7: Multi-Round Scoring Quiz (with Data Structures)
**Objective:** Combine loops, conditionals, error handling, and dictionaries to build a stateful application.

Upgrade Exercise 3 into a full multi-question quiz. The questions and answers are stored in a **list of dictionaries**. The program tracks the user's score across all questions and displays a final report card.

**Requirements:**
- Store at least 5 questions in a list of dicts: `[{"question": ..., "answer": ..., "hint": ...}, ...]`
- The user gets **2 attempts per question**. After the first wrong answer, display the hint.
- Clean all user input (`.strip().lower()`).
- At the end, print a summary: total score, percentage, and a grade (A/B/C/F).
- Use `try/except` if any question expects a numeric answer.

````python
quiz_bank = [
    {"question": "What keyword creates a function in Python?", "answer": "def", "hint": "It's a 3-letter word."},
    {"question": "What data type is immutable: list or tuple?", "answer": "tuple", "hint": "It uses parentheses."},
    {"question": "What is 2 ** 10?", "answer": "1024", "hint": "It's a power of 2."},
    {"question": "What method removes whitespace from both ends of a string?", "answer": "strip", "hint": "Think .___()."},
    {"question": "What built-in function returns the number of items in a list?", "answer": "len", "hint": "Short for length."},
]

def run_full_quiz(questions):
    score = 0
    total = len(questions)
    
    # TODO: Loop through each question dict
    # TODO: Give the user 2 attempts; show hint after the first failure
    # TODO: Clean user input before comparing
    # TODO: Track score
    for question in quiz_bank:
        attempts = 2

        while attempts > 0:
            answer = str(input(question["question"])).strip().lower()
            if answer == question["answer"]:
                score += 1
                print("Correct answer!")
                break
            attempts -= 1
            if attempts > 0:
                if attempts >0:
                    print(f"Hint: {question['hint']} You have one attempt left.")
                else:
                    print("No attempts left.")
    if score / total >=90:
        return "A"
    if 90 > score/total >= 70:
        return "B"
    if  70 > score/total >= 50:
        return "C"
    else :
        return "F"
    # TODO: Print final report card with score, percentage, and grade
    # Grading scale: >= 90% → A, >= 70% → B, >= 50% → C, else F

grade = run_full_quiz(quiz_bank)
print(grade)  # Uncomment to test
````

## Exercise 8: Expense Tracker with Category Reports
**Objective:** Build a complete menu-driven application that combines everything: loops, nested conditionals, error handling, functions, and data structures.

Create a mini expense tracker where the user can add expenses, view them by category, and see a summary report. This exercise ties together all the concepts from Session II.

**Menu Options:**
1. **Add Expense** — Ask for amount (float), category (string), and description (string). Validate that the amount is a positive number.
2. **View by Category** — Show all expenses for a user-selected category.
3. **Summary Report** — Print: total spent, number of transactions, average transaction, and the category with the highest spend.
4. **Quit**

**Data Structure:** Store expenses as a list of dictionaries:
```python
[{"amount": 12.50, "category": "food", "description": "Lunch"}, ...]
```

**Defensive Coding Requirements:**
- `try/except ValueError` for the amount input.
- Reject negative or zero amounts.
- Handle the case where the user asks for a summary with zero expenses (avoid `ZeroDivisionError`).

````python
expenses = []

def add_expense(expenses_list):
    """Prompt user for expense details and append to the list."""
    # TODO: Ask for amount with try/except for ValueError
    # TODO: Reject negative or zero amounts
    # TODO: Ask for category and description
    # TODO: Append a dict to expenses_list
    ammount = float(input("Please enter the ammount of the expense"))
    if ammount < 0:
        raise ValueError("Value is below zero")
    if ammount == 0:
        raise ValueError("Value is equal zero")
    category = input("Please enter the category")
    description = input("Please enter the description")
    expenses_list.append({
        "amount": ammount,
        "category": category,
        "description": description,
    })
        

def view_by_category(expenses_list):
    """Show expenses filtered by a user-selected category."""
    # TODO: Get unique categories from the list
    # TODO: Ask user to pick one
    # TODO: Loop and print matching expenses
    unique_categories = sorted({expense["category"] for expense in expenses_list})
    if not unique_categories:
        print("There are no expenses in the list yet")
        return
    for unique_category in unique_categories:
        print(f"Category name : {unique_category}" )
    picked_category = str(input("Please pick one of the categories")).strip()
    for expense_in_cat in expenses_list:
        if expense_in_cat["category"].strip().lower() == picked_category.lower():
            print(f"Amount: {expense_in_cat['amount']}, "
            f"description: {expense_in_cat['description']}")


def summary_report(expenses_list):
    """Print total, count, average, and top category."""
    # TODO: Handle empty list case (no ZeroDivisionError!)
    # TODO: Calculate total, count, average
    # TODO: Find the category with the highest total spend
    if not expenses_list:
        raise ZeroDivisionError("The list is currently empty")
    total = 0
    category_totals = {}
    for expense in expenses_list:
        amount = expense["amount"]
        category = expense["category"]

        total += amount
        category_totals[category] = category_totals.get(category, 0) + amount
    count = len(expenses_list)
    average = total/count
    top_category = max(category_totals, key = category_totals.get)
    print(f"Total: {total:.2f}")
    print(f"Number of expenses: {count}")
    print(f"Average: {average:.2f}")
    print(f"Highest-spending category: {top_category} "
          f"({category_totals[top_category]:.2f})")


def run_tracker():
    """Main menu loop."""
    # TODO: while True with menu: [1] Add [2] View by Category [3] Summary [4] Quit
    # TODO: Use conditionals to route to the right function
    # TODO: Handle invalid menu choices gracefully
    while True:
        choice = str(input("Pick one of the options : [1] Add [2] View by Category [3] Summary [4] Quit")).strip().lower()
        if choice == "q":
            print("Thank you and goodbye")
            break
        try:
            choice_int = int(choice)
        except:
            TypeError("Please enter one of the numbers that was offered")
            continue
        if choice_int <= 0 or choice_int > 4:
            raise ValueError("Please choose one of the numbers that was offered")
            continue
        if choice_int == 1:
            add_expense(expenses)
        if choice_int == 2:
            view_by_category(expenses)
        if choice_int == 3:
            summary_report(expenses)

run_tracker()  # Uncomment to test
````
