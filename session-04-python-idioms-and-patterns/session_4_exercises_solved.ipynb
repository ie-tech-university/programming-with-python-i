## Exercise 1: Loop Refactoring Workshop
**Objective:** Rewrite traditional `for` loops as list comprehensions.

You are given several functions that use the classic `for` loop + `.append()` pattern. Your job is to rewrite each one using a **single list comprehension** that produces the same result.

**Tasks:**
1. `get_squares(n)` → Return a list of squares from 1 to `n`.
2. `get_long_words(words, min_length)` → Return only words with length ≥ `min_length`, converted to uppercase.
3. `flatten(matrix)` → Flatten a list of lists into a single list.
4. `extract_emails(users)` → Given a list of user dicts, return a list of their email addresses.
5. `label_numbers(numbers)` → Return `"even"` or `"odd"` for each number (use ternary inside comprehension).

```python
# --- ORIGINAL LOOP VERSIONS (for reference) ---

def get_squares_loop(n):
    result = []
    for i in range(1, n + 1):
        result.append(i ** 2)
    return result

def get_long_words_loop(words, min_length):
    result = []
    for word in words:
        if len(word) >= min_length:
            result.append(word.upper())
    return result

def flatten_loop(matrix):
    result = []
    for row in matrix:
        for item in row:
            result.append(item)
    return result

def extract_emails_loop(users):
    result = []
    for user in users:
        result.append(user["email"])
    return result

def label_numbers_loop(numbers):
    result = []
    for n in numbers:
        if n % 2 == 0:
            result.append("even")
        else:
            result.append("odd")
    return result


# --- COMPREHENSION VERSIONS ---

def get_squares(n):
    """Return [1, 4, 9, ..., n²] using a list comprehension."""
    return [i ** 2 for i in range(1, n + 1)]

def get_long_words(words, min_length):
    """Return words with len >= min_length, converted to uppercase."""
    return [w.upper() for w in words if len(w) >= min_length]

def flatten(matrix):
    """Flatten a list of lists into a single list."""
    return [item for row in matrix for item in row]

def extract_emails(users):
    """Extract the 'email' field from each user dict."""
    return [user["email"] for user in users]

def label_numbers(numbers):
    """Return 'even' or 'odd' for each number using a ternary in a comprehension."""
    return ["even" if n % 2 == 0 else "odd" for n in numbers]


# --- TEST ---
print("Squares:", get_squares(5))
print("Long words:", get_long_words(["hi", "hello", "python", "go", "world"], 4))
print("Flatten:", flatten([[1, 2], [3, 4], [5, 6]]))

users = [{"name": "Alice", "email": "alice@x.com"}, {"name": "Bob", "email": "bob@x.com"}]
print("Emails:", extract_emails(users))
print("Labels:", label_numbers([1, 2, 3, 4, 5]))
```

## Exercise 2: Data Transformer
**Objective:** Practice dictionary comprehensions, set comprehensions, and data reshaping.

You are given a dataset of products. Write functions that use **dict and set comprehensions** to transform the data.

**Tasks:**
1. `apply_discount(products, pct)` → Return a new dict with all prices reduced by `pct`%.
2. `invert_dict(d)` → Swap keys and values.
3. `get_unique_categories(products)` → Return a set of all unique categories using a set comprehension.
4. `expensive_products(products, threshold)` → Return a dict of only products with price > threshold.
5. `category_index(products)` → Return a dict mapping each category to a list of product names in that category.

```python
products = {
    "Laptop": {"price": 999.99, "category": "Electronics"},
    "Headphones": {"price": 149.99, "category": "Electronics"},
    "Coffee Maker": {"price": 79.99, "category": "Kitchen"},
    "Notebook": {"price": 12.99, "category": "Office"},
    "Desk Lamp": {"price": 34.99, "category": "Office"},
    "Blender": {"price": 59.99, "category": "Kitchen"},
    "Tablet": {"price": 499.99, "category": "Electronics"},
}

def apply_discount(products, pct):
    """Return a dict {name: discounted_price} with pct% off each product."""
    return {name: round(info["price"] * (1 - pct / 100), 2) for name, info in products.items()}

def invert_dict(d):
    """Swap keys and values of a flat dictionary."""
    return {v: k for k, v in d.items()}

def get_unique_categories(products):
    """Return a set of all unique categories."""
    return {info["category"] for info in products.values()}

def expensive_products(products, threshold):
    """Return a dict of products with price > threshold."""
    return {name: info for name, info in products.items() if info["price"] > threshold}

def category_index(products):
    """Return {category: [product_names]} mapping."""
    index = {}
    for name, info in products.items():
        index.setdefault(info["category"], []).append(name)
    return index


# --- TEST ---
print("10% discount:", apply_discount(products, 10))

codes = {"ES": "Spain", "FR": "France", "DE": "Germany"}
print("Inverted:", invert_dict(codes))

print("Categories:", get_unique_categories(products))

print("Expensive (>100):", expensive_products(products, 100))

print("Category index:", category_index(products))
```

## Exercise 3: Parallel Data Merger
**Objective:** Use `zip()` and `enumerate()` to combine and align parallel data sources.

You have student data split across multiple separate lists (a common situation when importing from spreadsheets). Use `zip()` to merge them and `enumerate()` to add ranking.

**Tasks:**
1. `merge_student_data(names, ages, scores)` → Use `zip()` to return a list of dicts.
2. `create_roster(names, ids)` → Use `zip()` + dict comprehension to return `{id: name}`.
3. `rank_students(names, scores)` → Use `zip()` + `sorted()` + `enumerate()` to return ranked strings.
4. `pair_reviews(students)` → Pair each student with the next one (wrap around).

```python
names = ["Alice", "Bob", "Charlie", "Diana", "Eve"]
ages = [22, 25, 23, 21, 24]
scores = [92, 78, 85, 95, 88]
student_ids = ["S001", "S002", "S003", "S004", "S005"]

def merge_student_data(names, ages, scores):
    """Merge parallel lists into a list of dicts using zip()."""
    return [{"name": n, "age": a, "score": s} for n, a, s in zip(names, ages, scores)]

def create_roster(names, ids):
    """Create a {id: name} dict using zip() + dict comprehension."""
    return {sid: name for sid, name in zip(ids, names)}

def rank_students(names, scores):
    """Rank students by score (highest first). Return ['#1: Name (score)', ...]."""
    paired = sorted(zip(names, scores), key=lambda x: x[1], reverse=True)
    return [f"#{rank}: {name} ({score})" for rank, (name, score) in enumerate(paired, start=1)]

def pair_reviews(students):
    """Pair each student with the next one (last wraps to first)."""
    rotated = students[1:] + students[:1]
    return list(zip(students, rotated))


# --- TEST ---
print("Merged:", merge_student_data(names, ages, scores))

print("Roster:", create_roster(names, student_ids))

print("Rankings:")
for line in rank_students(names, scores):
    print(f"  {line}")

print("Peer review pairs:", pair_reviews(names))
```

## Exercise 4: Lambda Sorting Lab
**Objective:** Use `lambda` functions as sorting keys with `sorted()`.

You are given a dataset of employees. Write functions that sort the data in various ways using `sorted()` with `key=lambda`.

**Tasks:**
1. `sort_by_salary(employees)` → Sort ascending by salary.
2. `sort_by_name_length(employees)` → Sort by the length of the employee's name (shortest first).
3. `sort_by_department_then_salary(employees)` → Sort alphabetically by department, then by salary descending within each department.
4. `get_top_earners(employees, n)` → Return the top `n` highest-paid employees.
5. `sort_strings_naturally(items)` → Sort strings like `'item1'`, `'item2'`, `'item10'` in natural numeric order.

```python
employees = [
    {"name": "Alice", "department": "Engineering", "salary": 95000},
    {"name": "Bob", "department": "Marketing", "salary": 62000},
    {"name": "Charlie", "department": "Engineering", "salary": 88000},
    {"name": "Diana", "department": "HR", "salary": 71000},
    {"name": "Eve", "department": "Marketing", "salary": 58000},
    {"name": "Frank", "department": "Engineering", "salary": 102000},
    {"name": "Grace", "department": "HR", "salary": 67000},
]

def sort_by_salary(employees):
    """Sort by salary ascending."""
    return sorted(employees, key=lambda e: e["salary"])

def sort_by_name_length(employees):
    """Sort by length of name (shortest first)."""
    return sorted(employees, key=lambda e: len(e["name"]))

def sort_by_department_then_salary(employees):
    """Sort by department (A-Z), then salary (highest first) within each dept."""
    return sorted(employees, key=lambda e: (e["department"], -e["salary"]))

def get_top_earners(employees, n):
    """Return the top n highest-paid employees."""
    return sorted(employees, key=lambda e: e["salary"], reverse=True)[:n]

def sort_strings_naturally(items):
    """Sort strings like 'item1', 'item2', 'item10' in natural numeric order."""
    return sorted(items, key=lambda s: int(''.join(filter(str.isdigit, s))))


# --- TEST ---
print("By salary:")
for e in sort_by_salary(employees):
    print(f"  {e['name']}: ${e['salary']:,}")

print("\nBy name length:")
for e in sort_by_name_length(employees):
    print(f"  {e['name']}")

print("\nBy department then salary:")
for e in sort_by_department_then_salary(employees):
    print(f"  {e['department']}: {e['name']} (${e['salary']:,})")

print("\nTop 3 earners:")
for e in get_top_earners(employees, 3):
    print(f"  {e['name']}: ${e['salary']:,}")

items = ["item10", "item2", "item1", "item20", "item3"]
print("\nNatural sort:", sort_strings_naturally(items))
```

## Exercise 5: Functional Data Pipeline
**Objective:** Use `map()`, `filter()`, and `lambda` to build data processing pipelines.

You are given raw sales data as a list of dictionaries. Build a pipeline that cleans, filters, transforms, and summarizes the data using functional programming tools.

**Tasks:**
1. `clean_names(records)` → Use `map()` to strip and title-case product names.
2. `filter_by_min_quantity(records, min_qty)` → Use `filter()` to keep records with quantity ≥ `min_qty`.
3. `add_revenue_field(records)` → Use `map()` to add a `"revenue"` key to each record.
4. `get_product_names(records)` → Use `map()` to extract just the product names.
5. `pipeline(records, min_qty)` → Chain all the above together.

```python
raw_sales = [
    {"product": "  laptop  ", "price": 999.99, "quantity": 3},
    {"product": "MOUSE", "price": 29.99, "quantity": 15},
    {"product": "  Keyboard ", "price": 59.99, "quantity": 8},
    {"product": "monitor", "price": 349.99, "quantity": 2},
    {"product": " HEADSET ", "price": 89.99, "quantity": 1},
    {"product": "webcam", "price": 49.99, "quantity": 12},
    {"product": "  USB Hub  ", "price": 19.99, "quantity": 25},
]

def clean_names(records):
    """Use map() to strip and title-case each product name."""
    return list(map(
        lambda r: {**r, "product": r["product"].strip().title()},
        records
    ))

def filter_by_min_quantity(records, min_qty):
    """Use filter() to keep records with quantity >= min_qty."""
    return list(filter(lambda r: r["quantity"] >= min_qty, records))

def add_revenue_field(records):
    """Use map() to add a 'revenue' key to each record."""
    return list(map(
        lambda r: {**r, "revenue": round(r["price"] * r["quantity"], 2)},
        records
    ))

def get_product_names(records):
    """Use map() to extract just the product names."""
    return list(map(lambda r: r["product"], records))

def pipeline(records, min_qty):
    """Chain: clean → filter → add revenue → sort by revenue descending."""
    cleaned = clean_names(records)
    filtered = filter_by_min_quantity(cleaned, min_qty)
    with_revenue = add_revenue_field(filtered)
    return sorted(with_revenue, key=lambda r: r["revenue"], reverse=True)


# --- TEST ---
print("Cleaned names:", get_product_names(clean_names(raw_sales)))

print("\nFiltered (qty >= 5):", get_product_names(filter_by_min_quantity(raw_sales, 5)))

print("\nFull pipeline (min_qty=3):")
for item in pipeline(raw_sales, 3):
    print(f"  {item['product']}: {item['quantity']} units, ${item['revenue']:.2f} revenue")
```

## Exercise 6: Pythonic Refactoring Challenge (Capstone)
**Objective:** Refactor an entire working script from verbose, non-Pythonic code into clean, idiomatic Python.

The code below works correctly but is written in a **non-Pythonic** style. It uses manual counters instead of `enumerate()`, manual loops instead of comprehensions, no `zip()`, no `lambda`, and no ternary expressions.

**Your task:** Rewrite the `analyze_class_pythonic()` function to produce the **exact same output** but using:
- List comprehensions (at least 2)
- `zip()` (at least 1)
- `enumerate()` (at least 1)
- `sorted()` with `key=lambda` (at least 1)
- Ternary expressions (at least 1)
- `map()` or `filter()` (at least 1)

```python
# ============================================================
# NON-PYTHONIC VERSION (for reference)
# ============================================================

def analyze_class_verbose():
    names = ["Alice", "Bob", "Charlie", "Diana", "Eve"]
    math_scores = [88, 95, 72, 91, 84]
    science_scores = [92, 78, 85, 89, 95]

    records = []
    i = 0
    while i < len(names):
        record = {
            "name": names[i],
            "math": math_scores[i],
            "science": science_scores[i],
        }
        records.append(record)
        i = i + 1

    averages = []
    for record in records:
        avg = (record["math"] + record["science"]) / 2
        averages.append(avg)

    labels = []
    for avg in averages:
        if avg >= 80:
            labels.append("PASS")
        else:
            labels.append("FAIL")

    paired = []
    j = 0
    while j < len(names):
        paired.append((names[j], averages[j], labels[j]))
        j = j + 1

    for a in range(len(paired)):
        for b in range(a + 1, len(paired)):
            if paired[b][1] > paired[a][1]:
                temp = paired[a]
                paired[a] = paired[b]
                paired[b] = temp

    print("===== CLASS REPORT =====")
    rank = 1
    for item in paired:
        print(f"  #{rank}: {item[0]} — Avg: {item[1]:.1f} [{item[2]}]")
        rank = rank + 1

    passing = []
    for item in paired:
        if item[2] == "PASS":
            passing.append(item[0])

    print(f"\nPassing students: {passing}")
    print(f"Pass rate: {len(passing)}/{len(names)}")

analyze_class_verbose()
print()
```

```python
# ============================================================
# PYTHONIC VERSION — produces the exact same output
# ============================================================

def analyze_class_pythonic():
    names = ["Alice", "Bob", "Charlie", "Diana", "Eve"]
    math_scores = [88, 95, 72, 91, 84]
    science_scores = [92, 78, 85, 89, 95]

    # 1. zip() to combine parallel lists into records
    records = [{"name": n, "math": m, "science": s}
               for n, m, s in zip(names, math_scores, science_scores)]

    # 2. List comprehension for averages
    averages = [(r["math"] + r["science"]) / 2 for r in records]

    # 3. Ternary expression inside comprehension for labels
    labels = ["PASS" if avg >= 80 else "FAIL" for avg in averages]

    # 4. zip() to pair + sorted() with key=lambda
    paired = sorted(zip(names, averages, labels),
                    key=lambda x: x[1], reverse=True)

    # 5. enumerate(start=1) for ranked output
    print("===== CLASS REPORT =====")
    for rank, (name, avg, label) in enumerate(paired, start=1):
        print(f"  #{rank}: {name} — Avg: {avg:.1f} [{label}]")

    # 6. filter() for passing students
    passing = list(map(lambda x: x[0], filter(lambda x: x[2] == "PASS", paired)))

    print(f"\nPassing students: {passing}")
    print(f"Pass rate: {len(passing)}/{len(names)}")

analyze_class_pythonic()
```
