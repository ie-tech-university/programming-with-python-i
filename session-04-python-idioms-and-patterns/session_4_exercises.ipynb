## Exercise 1: Loop Refactoring Workshop
**Objective:** Rewrite traditional `for` loops as list comprehensions.

You are given several functions that use the classic `for` loop + `.append()` pattern. Your job is to rewrite each one using a **single list comprehension** that produces the same result.

**Tasks:**
1. `get_squares(n)` → Return a list of squares from 1 to `n`.
2. `get_long_words(words, min_length)` → Return only words with length ≥ `min_length`, converted to uppercase.
3. `flatten(matrix)` → Flatten a list of lists into a single list.
4. `extract_emails(users)` → Given a list of user dicts, return a list of their email addresses.
5. `label_numbers(numbers)` → Return `"even"` or `"odd"` for each number (use ternary inside comprehension).

```python
# --- ORIGINAL LOOP VERSIONS (DO NOT MODIFY) ---

def get_squares_loop(n):
    result = []
    for i in range(1, n + 1):
        result.append(i ** 2)
    return result

def get_long_words_loop(words, min_length):
    result = []
    for word in words:
        if len(word) >= min_length:
            result.append(word.upper())
    return result

def flatten_loop(matrix):
    result = []
    for row in matrix:
        for item in row:
            result.append(item)
    return result

def extract_emails_loop(users):
    result = []
    for user in users:
        result.append(user["email"])
    return result

def label_numbers_loop(numbers):
    result = []
    for n in numbers:
        if n % 2 == 0:
            result.append("even")
        else:
            result.append("odd")
    return result


# --- YOUR COMPREHENSION VERSIONS BELOW ---

def get_squares(n):
    return [number**2 for number in range(1, n+1)]

def get_long_words(words, min_length):
    return [word.upper() for word in words if len(word) >= min_length]

def flatten(matrix):
    return [item for row in matrix for item in row]

def extract_emails(users):
    return [item['email'] for item in users]

def label_numbers(numbers):
    return ['even' if number % 2 == 0 else 'odd' for number in numbers]

# --- TEST YOUR FUNCTIONS ---
print("Squares:", get_squares(5))
# Expected: [1, 4, 9, 16, 25]

print("Long words:", get_long_words(["hi", "hello", "python", "go", "world"], 4))
# Expected: ['HELLO', 'PYTHON', 'WORLD']

print("Flatten:", flatten([[1, 2], [3, 4], [5, 6]]))
# Expected: [1, 2, 3, 4, 5, 6]

users = [{"name": "Alice", "email": "alice@x.com"}, {"name": "Bob", "email": "bob@x.com"}]
print("Emails:", extract_emails(users))
# Expected: ['alice@x.com', 'bob@x.com']

print("Labels:", label_numbers([1, 2, 3, 4, 5]))
# Expected: ['odd', 'even', 'odd', 'even', 'odd']
```

## Exercise 2: Data Transformer
**Objective:** Practice dictionary comprehensions, set comprehensions, and data reshaping.

You are given a dataset of products. Write functions that use **dict and set comprehensions** to transform the data.

**Tasks:**
1. `apply_discount(products, pct)` → Return a new dict with all prices reduced by `pct`%.
2. `invert_dict(d)` → Swap keys and values.
3. `get_unique_categories(products)` → Return a set of all unique categories using a set comprehension.
4. `expensive_products(products, threshold)` → Return a dict of only products with price > threshold.
5. `category_index(products)` → Return a dict mapping each category to a list of product names in that category.

```python
products = {
    "Laptop": {"price": 999.99, "category": "Electronics"},
    "Headphones": {"price": 149.99, "category": "Electronics"},
    "Coffee Maker": {"price": 79.99, "category": "Kitchen"},
    "Notebook": {"price": 12.99, "category": "Office"},
    "Desk Lamp": {"price": 34.99, "category": "Office"},
    "Blender": {"price": 59.99, "category": "Kitchen"},
    "Tablet": {"price": 499.99, "category": "Electronics"},
}

def apply_discount(products, pct):
    return {name: round(info["price"] * (1 - pct/100), 2) for name, info in products.items()}

def invert_dict(d):
    return {value: key for key, value in d.items()}

def get_unique_categories(products):
    return {info["category"] for _, info in products.items()}

def expensive_products(products, threshold):
    return {name: product for name, product in products.items() if product['price'] > threshold}

def category_index(products):
    categories = get_unique_categories(products)
    cat_dicts = {cat: [] for cat in categories}
    for name, info in products.items():
        cat_dicts[info['category']].append(name)

    return cat_dicts

# --- TEST ---
print("10% discount:", apply_discount(products, 10))

codes = {"ES": "Spain", "FR": "France", "DE": "Germany"}
print("Inverted:", invert_dict(codes))

print("Categories:", get_unique_categories(products))

print("Expensive (>100):", expensive_products(products, 100))

print("Category index:", category_index(products))
```

## Exercise 3: Parallel Data Merger
**Objective:** Use `zip()` and `enumerate()` to combine and align parallel data sources.

You have student data split across multiple separate lists (a common situation when importing from spreadsheets). Use `zip()` to merge them and `enumerate()` to add ranking.

**Tasks:**
1. `merge_student_data(names, ages, scores)` → Use `zip()` to return a list of dicts: `[{"name": ..., "age": ..., "score": ...}, ...]`
2. `create_roster(names, ids)` → Use `zip()` + dict comprehension to return `{id: name}`.
3. `rank_students(names, scores)` → Use `zip()` + `sorted()` + `enumerate()` to return a list of strings like `"#1: Alice (95)"`.
4. `pair_reviews(students)` → Given a list of student names, pair each student with the next one for peer review. The last student pairs with the first (wrap around). Return a list of tuples.

```python
# Data split across separate lists
names = ["Alice", "Bob", "Charlie", "Diana", "Eve"]
ages = [22, 25, 23, 21, 24]
scores = [92, 78, 85, 95, 88]
student_ids = ["S001", "S002", "S003", "S004", "S005"]

def merge_student_data(names, ages, scores):
    return [{"name": name, "age": age, "score": score} for name, age, score in zip(names, ages, scores)]

def create_roster(names, ids):
    return { ide: name for ide, name in zip(ids, names)}

def rank_students(names, scores):
    ranked = sorted(zip(names, scores), key=lambda pair: pair[1], reverse=True)
    return [f"#{i}: {name} ({score})" for i, (name, score) in enumerate(ranked, start=1)]

def pair_reviews(students):
    return list(zip(students, students[1:] + students[:1]))



# --- TEST ---
print("Merged:", merge_student_data(names, ages, scores))

print("Roster:", create_roster(names, student_ids))

print("Rankings:")
for line in rank_students(names, scores):
    print(f"  {line}")

print("Peer review pairs:", pair_reviews(names))
# Expected: [('Alice', 'Bob'), ('Bob', 'Charlie'), ..., ('Eve', 'Alice')]
```

## Exercise 4: Lambda Sorting Lab
**Objective:** Use `lambda` functions as sorting keys with `sorted()`.

You are given a dataset of employees. Write functions that sort the data in various ways using `sorted()` with `key=lambda`.

**Tasks:**
1. `sort_by_salary(employees)` → Sort ascending by salary.
2. `sort_by_name_length(employees)` → Sort by the length of the employee's name (shortest first).
3. `sort_by_department_then_salary(employees)` → Sort alphabetically by department, then by salary descending within each department.
4. `get_top_earners(employees, n)` → Return the top `n` highest-paid employees.
5. `sort_strings_naturally(items)` → Given a list like `["item10", "item2", "item1"]`, sort them in natural numeric order (not alphabetical).

```python
employees = [
    {"name": "Alice", "department": "Engineering", "salary": 95000},
    {"name": "Bob", "department": "Marketing", "salary": 62000},
    {"name": "Charlie", "department": "Engineering", "salary": 88000},
    {"name": "Diana", "department": "HR", "salary": 71000},
    {"name": "Eve", "department": "Marketing", "salary": 58000},
    {"name": "Frank", "department": "Engineering", "salary": 102000},
    {"name": "Grace", "department": "HR", "salary": 67000},
]

def sort_by_salary(employees):
    return sorted(employees, key=lambda employee: employee['salary'])

def sort_by_name_length(employees):
    return sorted(employees, key=lambda employee: len(employee['name']))

def sort_by_department_then_salary(employees):
    return sorted(employees, key=lambda employee: (employee['department'], -employee['salary']))

def get_top_earners(employees, n):
    return sorted(employees, key=lambda employee: employee['salary'], reverse=True)[:n]

def sort_strings_naturally(items):
    """Sort strings like 'item1', 'item2', 'item10' in natural numeric order."""
    return sorted(items, key=lambda item: int(''.join(filter(str.isdigit, item))))


# --- TEST ---
print("By salary:")
for e in sort_by_salary(employees):
    print(f"  {e['name']}: ${e['salary']:,}")

print("\nBy name length:")
for e in sort_by_name_length(employees):
    print(f"  {e['name']}")

print("\nBy department then salary:")
for e in sort_by_department_then_salary(employees):
    print(f"  {e['department']}: {e['name']} (${e['salary']:,})")

print("\nTop 3 earners:")
for e in get_top_earners(employees, 3):
    print(f"  {e['name']}: ${e['salary']:,}")

items = ["item10", "item2", "item1", "item20", "item3"]
print("\nNatural sort:", sort_strings_naturally(items))
# Expected: ['item1', 'item2', 'item3', 'item10', 'item20']
```

## Exercise 5: Functional Data Pipeline
**Objective:** Use `map()`, `filter()`, and `lambda` to build data processing pipelines.

You are given raw sales data as a list of dictionaries. Build a pipeline that cleans, filters, transforms, and summarizes the data using functional programming tools.

**Tasks:**
1. `clean_names(records)` → Use `map()` to return a new list where each record's `"product"` name is stripped and title-cased.
2. `filter_by_min_quantity(records, min_qty)` → Use `filter()` to keep only records where `"quantity"` ≥ `min_qty`.
3. `add_revenue_field(records)` → Use `map()` to add a `"revenue"` key (`price × quantity`) to each record.
4. `get_product_names(records)` → Use `map()` to extract just the product names as a list of strings.
5. `pipeline(records, min_qty)` → Chain all the above: clean → filter → add revenue → sort by revenue descending. Return the final list.

```python
raw_sales = [
    {"product": "  laptop  ", "price": 999.99, "quantity": 3},
    {"product": "MOUSE", "price": 29.99, "quantity": 15},
    {"product": "  Keyboard ", "price": 59.99, "quantity": 8},
    {"product": "monitor", "price": 349.99, "quantity": 2},
    {"product": " HEADSET ", "price": 89.99, "quantity": 1},
    {"product": "webcam", "price": 49.99, "quantity": 12},
    {"product": "  USB Hub  ", "price": 19.99, "quantity": 25},
]

def clean_names(records):
    return map(lambda record: {**record, "product": record['product'].strip().title()}, records)

def filter_by_min_quantity(records, min_qty):
    return list(filter(lambda record: record['quantity'] >= min_qty, records))

def add_revenue_field(records):
    return map(lambda record: {**record, 'revenue': record['quantity'] * record['price']}, records)

def get_product_names(records):
    return list(map(lambda record: record['product'], records))

def pipeline(records, min_qty):
    cleaned_names = clean_names(records)
    filtered_by_qty = filter_by_min_quantity(cleaned_names, min_qty)
    with_revenue = add_revenue_field(filtered_by_qty)
    return sorted(with_revenue, key=lambda item: item['revenue'], reverse=True)

# --- TEST ---
print("Cleaned names:", get_product_names(clean_names(raw_sales)))

print("\nFiltered (qty >= 5):", get_product_names(filter_by_min_quantity(raw_sales, 5)))

print("\nFull pipeline (min_qty=3):")
for item in pipeline(raw_sales, 3):
    print(f"  {item['product']}: {item['quantity']} units, ${item['revenue']:.2f} revenue")
```
