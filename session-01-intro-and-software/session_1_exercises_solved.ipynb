## Exercise 3: Multi-Function Calculator
**Objective:** Write a complete calculator module using functions with type hints, docstrings, and default parameters.


```python
def add(a: float, b: float) -> float:
    """Return the sum of a and b."""
    return a + b

def subtract(a: float, b: float) -> float:
    """Return the difference of a and b."""
    return a - b

def multiply(a: float, b: float) -> float:
    """Return the product of a and b."""
    return a * b

def divide(a: float, b: float) -> float | None:
    """Return the quotient of a and b. Return None if b is zero."""
    if b == 0:
        return None
    return a / b

def power(base: float, exponent: float = 2) -> float:
    """Return base raised to the given exponent (default: squared)."""
    return base ** exponent

def calculator_report(a: float, b: float) -> str:
    """Return a formatted report of all four basic operations on a and b."""
    lines = [
        f"Calculator Report for {a} and {b}",
        f"{'='*35}",
        f"  {a} + {b} = {add(a, b)}",
        f"  {a} - {b} = {subtract(a, b)}",
        f"  {a} * {b} = {multiply(a, b)}",
    ]
    div_result = divide(a, b)
    if div_result is None:
        lines.append(f"  {a} / {b} = undefined (division by zero)")
    else:
        lines.append(f"  {a} / {b} = {div_result:.4f}")
    return "\n".join(lines)


# --- TEST ---
print(add(10, 3))
print(subtract(10, 3))
print(multiply(10, 3))
print(divide(10, 3))
print(divide(10, 0))
print(power(5))
print(power(2, 10))
print()
print(calculator_report(20, 4))
print()
print(calculator_report(10, 0))
```

## Exercise 4: Temperature Converter with Type Hints
**Objective:** Practice writing multiple small, focused functions that call each other.


```python
def celsius_to_fahrenheit(c: float) -> float:
    """Convert Celsius to Fahrenheit."""
    return c * 9 / 5 + 32

def fahrenheit_to_celsius(f: float) -> float:
    """Convert Fahrenheit to Celsius."""
    return (f - 32) * 5 / 9

def celsius_to_kelvin(c: float) -> float:
    """Convert Celsius to Kelvin."""
    return c + 273.15

def fahrenheit_to_kelvin(f: float) -> float:
    """Convert Fahrenheit to Kelvin by composing existing functions."""
    c = fahrenheit_to_celsius(f)
    return celsius_to_kelvin(c)

def temperature_table(start: int, end: int, step: int = 10) -> None:
    """Print a conversion table from start to end Celsius."""
    print(f"{'Celsius':>10} {'Fahrenheit':>12} {'Kelvin':>10}")
    print("-" * 35)
    for c in range(start, end + 1, step):
        f = celsius_to_fahrenheit(c)
        k = celsius_to_kelvin(c)
        print(f"{c:>10.1f} {f:>12.1f} {k:>10.2f}")


# --- TEST ---
print(celsius_to_fahrenheit(100))   # 212.0
print(fahrenheit_to_celsius(32))    # 0.0
print(celsius_to_kelvin(0))         # 273.15
print(fahrenheit_to_kelvin(212))    # 373.15
print()
temperature_table(0, 100, 20)
```

## Exercise 5: Student Grade Functions
**Objective:** Combine multiple concepts — functions, type hints, string formatting, and basic math — into a small grading utility.


```python
def calculate_average(scores: list[float]) -> float:
    """Return the average of a list of scores, or 0.0 if empty."""
    if not scores:
        return 0.0
    return sum(scores) / len(scores)

def letter_grade(average: float) -> str:
    """Return A/B/C/D/F based on the average score."""
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    return "F"

def student_summary(name: str, scores: list[float]) -> str:
    """Return a formatted summary string for one student."""
    avg = calculate_average(scores)
    grade = letter_grade(avg)
    return f"{name}: avg {avg:.2f} ({grade})"

def class_report(students: list[tuple[str, list[float]]]) -> None:
    """Print a formatted class report."""
    print("\n===== CLASS REPORT =====")
    print(f"{'Student':<12} {'Average':>8} {'Grade':>6}")
    print("-" * 28)

    averages = []
    for name, scores in students:
        avg = calculate_average(scores)
        grade = letter_grade(avg)
        averages.append((name, avg, grade))
        print(f"{name:<12} {avg:>8.2f} {grade:>6}")

    class_avg = sum(a for _, a, _ in averages) / len(averages)
    print("-" * 28)
    print(f"{'Class Avg':<12} {class_avg:>8.2f}")

    top_name, top_avg, top_grade = max(averages, key=lambda x: x[1])
    print(f"\nTop Student: {top_name} ({top_avg:.2f}, {top_grade})")


# --- TEST ---
students = [
    ("Alice", [90, 85, 92, 88]),
    ("Bob", [78, 81, 75, 80]),
    ("Charlie", [95, 98, 100, 97]),
    ("Diana", [60, 55, 70, 65]),
]

print(student_summary("Alice", [90, 85, 92, 88]))
print()
class_report(students)
```
